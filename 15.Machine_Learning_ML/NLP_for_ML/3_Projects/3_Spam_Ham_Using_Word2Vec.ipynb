````python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')  # WordNetLemmatizer ke liye zaruri hai
from nltk.stem import WordNetLemmatizer
import re
````

````python
lemmatizor = WordNetLemmatizer()
````

````python
corpus = [ ]

for i in range(0 , len(messages)):

    # 1st remove all the special charaters 

    review = re.sub('[^a-zA-Z]', ' ', messages["Message"][i])   # this means apart from small a-z if you find any other character subsititue it with blank ( " " )
    review = review.lower()

    # convert revirw convert it inro list of words
    review = review.split()

    # apply same process that we did in stemming .

    review = [lemmatizor.lemmatize(word) for word in review ]

    review = ' '.join(review)

    corpus.append(review)


````

## Parameters

| Param | Default | Meaning |
|-------|---------|---------|
| `sentences` | required | list of lists of words |
| `vector_size` | 100 | vector dimensions per word |
| `window` | 5 | max distance between target & context word |
| `min_count` | 5 | ignore words with freq < this |
| `workers` | 3 | CPU threads for training |
| `sg` | 0 | 0=CBOW, 1=Skip-gram |
| `hs` | 0 | 0=negative sampling, 1=hierarchical softmax |
| `negative` | 5 | negative samples per positive (if hs=0) |
| `epochs` | 5 | passes over corpus |
| `alpha` | 0.025 | initial learning rate |
| `min_alpha` | 0.0001 | final learning rate |
| `seed` | 1 | random seed |

---

## CBOW vs Skip-gram

| | CBOW | Skip-gram |
|-|------|-----------|
| `sg` | 0 | 1 |
| Input | context → predict word | word → predict context |
| Speed | Fast | Slow |
| Best for | frequent words | rare words |

# Average Word2Vec

Sometimes a sentence contains multiple words:

```text
"I love machine learning"
```

But a Machine Learning model may need **one fixed-size vector representing the entire sentence**.

Word2Vec gives us:

```text
I       → vector
love    → vector
machine → vector
learning→ vector
```

One simple approach is to **average the vectors**.

This is called:

> **Average Word2Vec**

---

#  Average Word2Vec Intuition

Suppose:

```text
I      → [1, 2, 3]
love   → [2, 4, 6]
Python → [3, 6, 9]
```

Average them dimension by dimension:

```text
Dimension 1:

(1 + 2 + 3) / 3
= 2
```

```text
Dimension 2:

(2 + 4 + 6) / 3
= 4
```

```text
Dimension 3:

(3 + 6 + 9) / 3
= 6
```

Therefore:

```text
Sentence vector:

[2, 4, 6]
```

---

# Average Word2Vec Formula

If a sentence contains `n` word vectors:

```text
v₁, v₂, v₃, ..., vₙ
```

then:

```text
Sentence Vector =
(v₁ + v₂ + v₃ + ... + vₙ) / n
```

The result has the **same number of dimensions** as the original Word2Vec vectors.

For example:

```text
Word vectors = 100 dimensions

        ↓ average

Sentence vector = 100 dimensions
```


````python
# Function: ek document ke saare words ke word2vec vectors ka average nikalta hai
def avg_word2vec(doc):
    
    # Step 1: Har word ka vector uthao (sirf jo model ki vocabulary mein hain)
    # - model.wv[word] → us word ka 100-dim vector
    # - if word in model.wv.index_to_key → OOV words skip ho jayenge
    vectors = [model.wv[word] for word in doc if word in model.wv.index_to_key]
    
    # Step 2: AGR KOI VECTOR OUT OF VOCABULARY H (saare words unknown the)
    # - empty list hogi to np.mean crash karega
    # - isliye zero vector return karo same size ka
    if len(vectors) == 0:
        return np.zeros(model.vector_size)  # fallback for OOV sentences
    
    # Step 3: Sab vectors ka column-wise mean nikalo
    # - axis=0 → har dimension ka alag alag average
    # - result → ek single vector jo poore sentence ka matlab represent karta hai
    return np.mean(vectors, axis=0)

from tqdm import tqdm

# Har sentence ko ek avg vector mein convert karo
X = []
for i in tqdm(range(len(words))):    # words = list of tokenized sentences
    X.append(avg_word2vec(words[i])) # har sentence → 1 avg vector → X mein append
````
