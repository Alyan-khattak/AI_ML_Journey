# Gradient Boosting (Theory)

Gradient Boosting is a **Supervised Ensemble Learning Algorithm** based on the **Boosting** technique.

It builds **multiple weak learners (usually Decision Trees)** **sequentially**, where each new tree learns from the mistakes (errors/residuals) made by the previous trees.

Unlike Random Forest (Bagging), where all trees are built independently, Gradient Boosting builds one tree at a time.

---

# Intuition

Imagine a teacher checking your exam.

```
Exam

↓

Teacher finds mistakes

↓

You correct mistakes

↓

Teacher checks again

↓

Fewer mistakes

↓

Repeat
```

Every correction makes your final result better.

Gradient Boosting works exactly the same way.

---

# Why Do We Need Gradient Boosting?

Suppose a single Decision Tree predicts house prices.

```
Actual

100
120
150
200
```

Predicted

```
110
125
140
180
```

Errors

```
-10
-5
10
20
```

Instead of building another independent tree,

Gradient Boosting asks:

> **Can the next tree learn these errors?**

If yes,

the prediction becomes more accurate.

---

# Basic Flow

```
Training Data
      │
      ▼
Base Prediction
      │
Residuals
      │
      ▼
Tree 1 learns residuals
      │
New Prediction
      │
Residuals
      │
      ▼
Tree 2 learns new residuals
      │
      ▼
...
      │
      ▼
Final Prediction
```

---

# Step 1 — Build the Base Model

For regression,

the first prediction is simply the **mean of the target values**.

Example

```
Actual Prices

100
120
150
130
```

Mean

```
(100 + 120 + 150 + 130) / 4

=

125
```

Initial Prediction

```
125
125
125
125
```

This is called the **Base Model (F₀)**.

```
F₀(x) = Mean(y)
```

---

# Step 2 — Compute Residuals

Residual means

```
Residual = Actual − Prediction
```

Formula

$$
r_i = y_i - \hat{y}_i
$$

Example

| Actual | Prediction | Residual |
|--------|-----------|----------|
|100|125|-25|
|120|125|-5|
|150|125|25|
|130|125|5|

Residuals tell us **how wrong** the model is.

---

# Step 3 — Train First Decision Tree

Now create a small Decision Tree.

But notice

The target is **NOT y**.

The target is

```
Residuals
```

Training Data

| X | Residual |
|---|----------|
|x₁|-25|
|x₂|-5|
|x₃|25|
|x₄|5|

Tree 1 learns

```
Residuals
```

instead of prices.

---

# Step 4 — Predict Residuals

Suppose Tree 1 predicts

```
-20

-4

20

8
```

These are **estimated residuals**.

Let's call them

```
R₁(x)
```

---

# Step 5 — Update Predictions

New Prediction

```
Old Prediction

+

Learning Rate

×

Residual Prediction
```

Formula

$$
F_1(x)=F_0(x)+\eta R_1(x)
$$

where

- \(F_0(x)\) = Base Prediction
- \(R_1(x)\) = Prediction from Tree 1
- \(\eta\) = Learning Rate

Example

Learning Rate

```
0.1
```

Old Prediction

```
125
```

Residual Prediction

```
20
```

New Prediction

```
125 + (0.1 × 20)

=

127
```

Notice

We do **not completely trust** Tree 1.

We move only a small step.

---

# Step 6 — Compute New Residuals

Again

```
Residual

=

Actual

-

New Prediction
```

Formula

$$
r_i^{(2)} = y_i - F_1(x_i)
$$

Residuals become smaller.

---

# Step 7 — Train Tree 2

Tree 2 learns

```
New Residuals
```

Again,

not the original target.

---

# Step 8 — Update Again

Formula

$$
F_2(x)=F_1(x)+\eta R_2(x)
$$

Prediction keeps improving.

---

# Continue Until

Repeat

```
Residual

↓

Decision Tree

↓

Update Prediction
```

until

- Maximum trees reached
- Residuals become very small

---

# Visual Flow

```
               Mean(y)
                  │
                  ▼
            Base Prediction
                  │
          Compute Residuals
                  │
                  ▼
         Decision Tree 1
                  │
                  ▼
        Residual Prediction
                  │
                  ▼
      Update Prediction (F₁)
                  │
                  ▼
        Compute New Residuals
                  │
                  ▼
         Decision Tree 2
                  │
                  ▼
      Update Prediction (F₂)
                  │
                  ▼
                Repeat
```

---

# Mathematical Formula

Initial Model

$$
F_0(x)=\bar{y}
$$

Residual

$$
r_i=y_i-F_{m-1}(x_i)
$$

Update

$$
F_m(x)=F_{m-1}(x)+\eta h_m(x)
$$

where

- \(h_m(x)\) = prediction from the new tree
- \(\eta\) = learning rate

# XGBoost (Extreme Gradient Boosting)

XGBoost stands for **Extreme Gradient Boosting**.

It is an **optimized implementation of Gradient Boosting** that is faster, more accurate, and includes built-in regularization to reduce overfitting.

It was designed by **Tianqi Chen** and has become one of the most widely used machine learning algorithms for structured (tabular) data.

---

# Intuition

Gradient Boosting says:

```
Current Prediction

↓

Find Errors (Residuals)

↓

Build New Tree

↓

Correct Errors
```

XGBoost follows the **same idea**, but it improves every step.

It asks:

- Which split reduces the error the most?
- Is adding this tree worth the extra complexity?
- Should we prune this branch?
- How confident are we in each leaf?

---

# Why Do We Need XGBoost?

Gradient Boosting is powerful but has limitations.

```
Gradient Boosting

↓

Slower

Can Overfit

No Regularization

Less Optimized
```

XGBoost solves these problems by adding:

- Regularization
- Better split selection
- Faster computation
- Parallel processing
- Automatic pruning
- Missing value handling

---

# Step 1 — Build the Base Model

## Regression

Mean

```
125
```

Initial Prediction

```
125

125

125

125
```

## Classification

Instead of predicting

```
0

or

1
```

XGBoost predicts **Log Odds**.

---

# What are Log Odds?

Probability

```
P = 0.50
```

Odds

$$
Odds=\frac{P}{1-P}
$$

Example

```
P = 0.8
```

Odds

$$
\frac{0.8}{0.2}=4
$$

Meaning

```
4 : 1
```

---

Log Odds

$$
\log\left(\frac{P}{1-P}\right)
$$

Example

```
P = 0.5
```

$$
\log\left(\frac{0.5}{0.5}\right)=0
$$

Therefore,

the initial prediction for binary classification is often

```
0
```

which corresponds to

```
Probability = 0.5
```

---

# Step 2 — Compute Gradients

Gradient means

> **How wrong is the current prediction?**

Formula

$$
g_i=\frac{\partial L}{\partial \hat{y_i}}
$$

where

- \(L\) = Loss Function
- \(g_i\) = First Derivative (Gradient)

The gradient tells the model **which direction to move** to reduce the loss.

---

# Step 3 — Compute Hessians

Unlike Gradient Boosting,

XGBoost also computes the **Second Derivative**.

Formula

$$
h_i=\frac{\partial^2 L}{\partial \hat{y_i}^2}
$$

This measures

```
How quickly the gradient is changing.
```

Think of it like driving a car.

Gradient

```
Which direction?
```

Hessian

```
How sharply should I turn?
```

Using both allows XGBoost to make smarter updates.

---

# Step 4 — Build a Decision Tree

The tree is built using

```
Features

+

Gradients

+

Hessians
```

instead of just the residuals.

Example

```
Age ≤ 40?

      Yes       No
```

---

# Step 5 — Compute Similarity Score

Each node receives a **Similarity Score**.

This tells us

```
How good is this node?
```

Formula

$$
Similarity=\frac{(\sum g)^2}{\sum h+\lambda}
$$

where

- \(\sum g\) = Sum of Gradients
- \(\sum h\) = Sum of Hessians
- \(\lambda\) = Regularization Parameter

Large Similarity Score

↓

Good Node

Small Score

↓

Poor Node

---

# Step 6 — Compute Gain

Now XGBoost asks

> Should this node be split?

Formula

$$
Gain=
\frac{1}{2}
\left(
Score_{Left}
+
Score_{Right}
-
Score_{Parent}
\right)
-
\gamma
$$

where

- Left Score
- Right Score
- Parent Score
- \(\gamma\) = Minimum gain required to split

---

Interpretation

```
Gain > 0

↓

Split
```

```
Gain ≤ 0

↓

Do Not Split
```

---

# Step 7 — Compute Leaf Weight

Once a leaf is created,

its prediction is

$$
w=
-\frac{\sum g}
{\sum h+\lambda}
$$

Notice

Decision Tree predicts

```
Average
```

Gradient Boosting predicts

```
Residual
```

XGBoost predicts

```
Leaf Weight
```

---

# Step 8 — Update Prediction

Formula

$$
F_m(x)
=
F_{m-1}(x)
+
\eta
w
$$

where

- \(w\) = Leaf Weight
- \(\eta\) = Learning Rate

Prediction gradually improves.

---

# Step 9 — Compute New Gradients

After updating,

XGBoost computes

```
New Gradients

New Hessians
```

Then builds another tree.

---

# Repeat

```
Prediction

↓

Gradient

↓

Hessian

↓

Decision Tree

↓

Similarity Score

↓

Gain

↓

Leaf Weight

↓

Update Prediction

↓

Repeat
```

# Interview Question

### Why is XGBoost better than Gradient Boosting?

Because XGBoost improves Gradient Boosting by using **first- and second-order derivatives (gradients and Hessians), regularization, gain-based split selection, automatic pruning, and several engineering optimizations that make it faster, more accurate, and less prone to overfitting.**

# One-Line Definition

> **XGBoost (Extreme Gradient Boosting) is an optimized boosting algorithm that builds decision trees sequentially using gradients and Hessians to minimize a regularized objective function, producing highly accurate and efficient predictive models.**
