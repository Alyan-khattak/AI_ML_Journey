# Stemming in NLP

## 1. What is Stemming?

> Root Word/ Stem Word

**Stemming** is the process of reducing a word to its **root/base form** by removing prefixes or suffixes.

Example:

```text
playing  → play
played   → play
plays    → play
```

The result is called a **stem**.

> Stemming is a simple, rule-based technique. The resulting stem **does not always have to be a real English word**.

---

# 2. Why Do We Use Stemming?

Different words can have similar meanings.

For example:

```text
play
playing
played
plays
```

For an NLP model, we may want to treat these as related words.

Instead of learning:

```text
play
playing
played
plays
```

we can reduce them to something closer to:

```text
play
play
play
play
```

This can reduce the number of unique words in our vocabulary.

---

# 3. Basic Idea

```text
            WORD
              ↓
       Remove suffix/prefix
              ↓
            STEM
```

Example:

```text
connected
    ↓
connect
```

```text
studies
    ↓
studi
```

Notice:

```text
"studi"
```

is **not a proper English word**.

That's completely normal with stemming.

---

# 4. Stemming in NLTK

NLTK provides several stemmers.

The most commonly used one for learning is:

```python

```

Import it:

```python
from nltk.stem import PorterStemmer
```

---

# 5. Basic Syntax

```python
stemmer = PorterStemmer()

stemmer.stem("playing")
```

Output:

```text

```

---

# 6. Example

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = [
    "playing",
    "played",
    "plays",
    "playing",
    "connection",
    "connected"
]

for word in words:
    print(word, "→", stemmer.stem(word))
```

Possible output:

```text
playing → play
played → play
plays → play
playing → play
connection → connect
connected → connect
```

# 7. Stemming a Sentence

First tokenize the sentence:

```python
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

text = "The students are studying and playing games."

words = word_tokenize(text)

stemmer = PorterStemmer()

stemmed_words = []

for word in words:
    stemmed_words.append(stemmer.stem(word))

print(stemmed_words)
```

Possible output:

```python
[
    "the",
    "student",
    "are",
    "studi",
    "and",
    "play",
    "game",
    "."
]
```

Notice:

```text
students → student
studying → studi
playing  → play
games    → game
```

---

# 8. Short Version

You can use list comprehension:

```python
stemmed_words = [
    stemmer.stem(word)
    for word in words
]

print(stemmed_words)
```

---

# 12. Stemming vs Normal Word

Stemming can produce a word that **doesn't exist**.

Example:

```text
studying → studi
```

This is because the stemmer mainly follows **rules**, rather than understanding the actual meaning of the word.

---

# 13. Stemming vs Lemmatization

### Stemming

```text
studies
   ↓
studi
```

Fast but sometimes crude.

### Lemmatization

```text
studies
   ↓
study
```

Tries to return a proper dictionary/base word.

```text
Stemming
   ↓
Fast + simple + sometimes inaccurate

Lemmatization
   ↓
More linguistically meaningful + usually slower
```

---

# 14. When Do We Use Stemming?

Stemming can be useful when:

- You want simple text normalization
- Speed is important
- Exact linguistic correctness is not critical
- You want to reduce vocabulary size
- Working with search engines or basic NLP pipelines

---

# RegexStemmer in NLTK

`RegexpStemmer` is an NLTK stemmer that uses **regular expressions (Regex)** to remove specific patterns from words.

Unlike `PorterStemmer`, where predefined stemming rules are used, with `RegexpStemmer` **you decide what pattern should be removed**.

---

## 1. Why Use RegexpStemmer?

Use it when you want **custom control** over stemming.

For example, you can tell it:

```text
Remove "ing" from the end of words
```

Then:

```text
playing → play
running → runn
working → work
```

---

## 2. Basic Syntax

```python
from nltk.stem import RegexpStemmer

stemmer = RegexpStemmer("ing$")

stemmer.stem("playing")
```

Output:

```text
play
```

---

## 3. What Does `"ing$"` Mean?

```text
ing$
```

Here:

```text
ing
```

means we are looking for `"ing"`.

And:

```text
$
```

means **at the end of the word**.

So:

```text
playing
       ↑
      ing
       ↑
      end
```

Therefore:

```text
playing → play
```

---

## 4. Example

```python
from nltk.stem import RegexpStemmer

stemmer = RegexpStemmer("ing$")

words = [
    "playing",
    "running",
    "working",
    "walking"
]

for word in words:
    print(word, "→", stemmer.stem(word))
```

Output:

```text
playing → play
running → runn
working → work
walking → walk
```

Notice:

```text
running → runn
```

It doesn't understand English.

It simply follows your Regex rule.

---

## 5. Custom Pattern

You can create your own pattern.

Example: remove `"ed"` from the end:

```python
stemmer = RegexpStemmer("ed$")

print(stemmer.stem("played"))
print(stemmer.stem("worked"))
print(stemmer.stem("walked"))
```

Output:

```text
play
work
walk
```

---

## 6. Removing Multiple Patterns

You can use Regex patterns such as:

```python
stemmer = RegexpStemmer(r"(ing|ed|s)$")
```

This means:

```text
ing OR ed OR s
```

at the end of the word.

Example:

```python
words = [
    "playing",
    "played",
    "plays",
    "working",
    "worked"
]

for word in words:
    print(word, "→", stemmer.stem(word))
```

---

## 7. When Should You Use RegexpStemmer?

Use it when:

- You want **custom stemming rules**
- You know exactly what patterns you want to remove
- You are working with a specific type of text
- You don't want the predefined rules of Porter/Snowball

### Example

If your dataset contains lots of words ending in:

```text
-ing
-ed
-s
```

you can create your own Regex rule.

---

## 8. PorterStemmer vs RegexpStemmer

| Stemmer | How it works |
|---|---|
| `PorterStemmer` | Uses predefined English stemming rules |
| `RegexpStemmer` | You define the Regex rules |
| `SnowballStemmer` | More advanced predefined rules |

### Simple Mental Model

```text
PorterStemmer
      ↓
"Use my built-in rules"

RegexpStemmer
      ↓
"Tell me your rules"
```

---

## Key Point

> **RegexpStemmer lets you create your own stemming rules using Regular Expressions.**

````python
from nltk.stem import RegexpStemmer
````

````python
words = [
    "playing",
    "running",
    "working",
    "walking"
]
````

````python
reg_stemer = RegexpStemmer(r"ing$")

for word in words:
    print(word + "  ---> ", reg_stemer.stem(word))
````

````python
words = [
    "playing",
    "played",
    "plays",
    "working",
    "worked",
    "cats"
]

````

````python
reg_stemer = RegexpStemmer(r"ing$|ed$|s$") # remove these suffix

for word in words:
     print(word + "  ---> ", reg_stemer.stem(word))
````

# Snowball Stemmer

`SnowballStemmer` is an NLTK stemmer used to reduce words to their **root/base stem**.

It is an improved and more flexible version of the classic `PorterStemmer`.

---

## 1. Import

```python
from nltk.stem import SnowballStemmer
```

---

## 2. Basic Syntax

```python
stemmer = SnowballStemmer("english")
```

Then:

```python
stemmer.stem("playing")
```

Output:

```text
play
```

---

## 3. Example

```python
from nltk.stem import SnowballStemmer

stemmer = SnowballStemmer("english")

words = [
    "playing",
    "played",
    "plays",
    "running",
    "studies",
    "connected"
]

for word in words:
    print(word, "→", stemmer.stem(word))
```

Possible output:

```text
playing → play
played → play
plays → play
running → run
studies → studi
connected → connect
```

---

## 4. Why Use SnowballStemmer?

It is useful when you want:

- Fast stemming
- Better stemming rules than classic Porter stemming
- Support for multiple languages
- A simple way to normalize related words

---

## 5. Multiple Languages

One advantage of Snowball Stemmer is that it supports multiple languages.

Example:

```python
SnowballStemmer("english")
```

```python

from nltk.stem import SnowballStemmer
```

---

## 8. Porter vs Snowball

| PorterStemmer | SnowballStemmer |
|---|---|
| Older algorithm | Improved algorithm |
| English-focused | Supports multiple languages |
| Simple | More refined rules |
| Commonly taught | Commonly used |

Example:

```python
from nltk.stem import PorterStemmer
from nltk.stem import SnowballStemmer

porter = PorterStemmer()
snowball = SnowballStemmer("english")

word = "running"

print("Porter:", porter.stem(word))
print("Snowball:", snowball.stem(word))
```

---

## 9. Important

Snowball is still **stemming**, not lemmatization.

It can produce a stem that isn't a proper English word:

```text
studies → studi
```

It does not fully understand the meaning of the word.

---

## Mental Model

```text
               WORD
                 ↓
        SnowballStemmer
                 ↓
       Apply stemming rules
                 ↓
               STEM
```

### Remember

> **SnowballStemmer is a rule-based stemmer that provides improved stemming algorithms and supports multiple languages.**

````python
from nltk.stem import SnowballStemmer

# Create stemmer
stemmer = SnowballStemmer("english")

# Words
words = [
    "playing",
    "played",
    "plays",
    "running",
    "runs",
    "studies",
    "studying",
    "connected",
    "connection",
    "happiness"
]

# Apply stemming
for word in words:
    print(word, "→", stemmer.stem(word))
````
