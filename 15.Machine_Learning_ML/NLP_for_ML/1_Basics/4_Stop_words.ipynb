# Stop Words in NLP

## 1. What Are Stop Words?

**Stop words** are common words that often provide little useful information for an NLP task.

Examples:

```text
the
is
a
an
and
or
in
on
of
to
for
```

Example sentence:

```text
"The cat is sitting on the table."
```

Possible stop words:

```text
The
is
on
the
```

After removing them:

```text
"cat sitting table"
```

---

# 2. Why Remove Stop Words?

Stop-word removal can:

- Reduce the number of words the model needs to process
- Reduce vocabulary size
- Remove words that may not help a particular task
- Make text processing faster

```text
Original text
      ↓
Tokenization
      ↓
Remove stop words
      ↓
Important words
```

# 5. Remove Stop Words

Example:

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

text = "I am learning Natural Language Processing."

words = word_tokenize(text)

stop_words = set(stopwords.words("english"))

filtered_words = []

for word in words:
    if word.lower() not in stop_words:
        filtered_words.append(word)

print(filtered_words)
```

Possible output:

```python
[
    "learning",
    "Natural",
    "Language",
    "Processing",
    "."
]
```

---

# 6. Short Version

Using list comprehension:

```python
filtered_words = [
    word
    for word in words
    if word.lower() not in stop_words
]

print(filtered_words)
```

# 7. Why Use `lower()`?

Consider:

```text
The
the
THE
```

NLTK's stop-word list contains lowercase words.

So:

```python
word.lower()
```

converts:

```text
"The" → "the"
"THE" → "the"
```

Then we can correctly check:

```python
word.lower() not in stop_words
```

# 8. Stop Words + Tokenization

A common basic NLP pipeline:

```text
Raw Text
   ↓
Tokenization
   ↓
Tokens
   ↓
Remove Stop Words
   ↓
Filtered Tokens
```

Example:

```text
"I am learning machine learning."

        ↓

["I", "am", "learning", "machine", "learning", "."]

        ↓

["learning", "machine", "learning", "."]
```

---

# 9. Important: Don't Always Remove Stop Words

Stop words are **not automatically useless**.

For example:

```text
"This movie is good."
```

and:

```text
"This movie is not good."
```

If we remove:

```text
"is"
"not"
```

we could lose important meaning.

So stop-word removal depends on the **NLP task**.

---

# 10. Stop Words vs Punctuation

Stop words:

```text
the
is
a
and
to
```

Punctuation:

```text
.
,
!
?
;
```

They are different things.

If you want to remove both, handle them separately.

# 11. Complete Example

```python
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download("stopwords")
nltk.download("punkt")
nltk.download("punkt_tab")

text = "I am learning Natural Language Processing with Python."

# Tokenization
words = word_tokenize(text)

# Stop words
stop_words = set(stopwords.words("english"))

# Remove stop words
filtered_words = [
    word
    for word in words
    if word.lower() not in stop_words
]

print("Original:")
print(words)

print("\nAfter removing stop words:")
print(filtered_words)
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import Porterstem

sentences = sent_tokenize(paragraph)

stemmer = PorterStemmer()

# list all the stop words
stop_words = set(stopwords.words("english"))

for i in range(len(sentences)):

    # 1. Sentence → words
    words = word_tokenize(sentences[i])

    # 2. Remove stop words
    filtered_words = []

    for word in words:
        if word.lower() not in stop_words:
            filtered_words.append(word)

    # 3. Lemmatization
    stem_words = []

    for word in stem_words:
        stem_words.append(stemmer.stem(word))

    # 4. Words → sentence
    sentences[i] = " ".join(stem_words)
