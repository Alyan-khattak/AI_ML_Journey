# Docker – Push an Image to Docker Hub

## 1. Start Docker

Before doing anything, make sure the Docker daemon is running.

For Docker Desktop on Linux:

```bash
systemctl --user start docker-desktop
```

Check that Docker is running:

```bash
docker info
```

If you see a `Server:` section, Docker is running.

---

## 2. Build the Docker Image

First, build your image from the directory containing your `Dockerfile`.

```bash
docker build -t welcome-app:latest .
```

### Breakdown

```text
docker build
```

Build a Docker image.

```text
-t welcome-app:latest
```

Give the image a name and tag.

```text
.
```

Use the current directory as the build context.

You can check the image:

```bash
docker images
```

You should see:

```text
welcome-app    latest    ...
```

---

## 3. Login to Docker Hub

Login to your Docker Hub account:

```bash
docker login
```

Docker may give you a device code and ask you to authenticate through your browser.

After successful authentication:

```text
Login Succeeded
```

### Important

Docker needs to store your credentials locally.

On Linux, this may require a configured credential store such as `pass` + GPG.

If you get:

```text
pass not initialized
```

initialize `pass` with your GPG key:

```bash
pass init YOUR_GPG_KEY_ID
```

Then login again:

```bash
docker login
```

---

# 4. Tag the Image

Your local image might currently be:

```text
welcome-app:latest
```

Docker Hub needs the image to be tagged with your Docker Hub username.

The format is:

```bash
docker tag <local-image>:<tag> <dockerhub-username>/<repository>:<tag>
```

For example:

```bash
docker tag welcome-app:latest alyanktk/welcome-app:latest
```

Now check:

```bash
docker images
```

You should see something similar to:

```text
REPOSITORY              TAG       IMAGE ID
welcome-app             latest    5bf45c1b6e30
alyanktk/welcome-app    latest    5bf45c1b6e30
```

Both images have the **same IMAGE ID**.

The second one is simply another tag pointing to the same image.

---

# 5. Push the Image

Now push the Docker image to Docker Hub:

```bash
docker push alyanktk/welcome-app:latest
```

The general syntax is:

```bash
docker push <dockerhub-username>/<repository>:<tag>
```

Example:

```bash
docker push alyanktk/welcome-app:latest
```

Docker will upload the image layers to Docker Hub.

You may see:

```text
The push refers to repository [docker.io/alyanktk/welcome-app]

86e037ebf8b6: Pushed
4f4fb700ef54: Pushed
1da342a38765: Pushed
...
latest: digest: sha256:...
```

---

# 6. Why Do We Need to Tag the Image?

If you run:

```bash
docker push welcome-app:latest
```

Docker interprets it as:

```text
docker.io/library/welcome-app
```

The `library` namespace is generally for official Docker images.

You don't own that namespace, so you may get:

```text
push access denied
```

Instead:

```bash
docker tag welcome-app:latest alyanktk/welcome-app:latest
```

Now Docker knows:

```text
Docker Hub
    │
    └── alyanktk
          │
          └── welcome-app
```

So:

```bash
docker push alyanktk/welcome-app:latest
```

pushes to your Docker Hub namespace.

---

# 7. Complete Workflow

From beginning to end:

```bash
# Start Docker Desktop
systemctl --user start docker-desktop

# Check Docker
docker info

# Build the image
docker build -t welcome-app:latest .

# Check images
docker images

# Login to Docker Hub
docker login

# Tag the image
docker tag welcome-app:latest alyanktk/welcome-app:latest

# Check the tag
docker images

# Push to Docker Hub
docker push alyanktk/welcome-app:latest
```

---

# 8. Understanding the Whole Process

The process is:

```text
                Local Computer
                     │
                     │
                Dockerfile
                     │
                     ↓
              docker build
                     │
                     ↓
             Docker Image
          welcome-app:latest
                     │
                     │ docker tag
                     ↓
       alyanktk/welcome-app:latest
                     │
                     │ docker push
                     ↓
                Docker Hub
                     │
                     ↓
       alyanktk/welcome-app
```

---

# 9. Pull the Image Later

Once the image is on Docker Hub, another computer can download it using:

```bash
docker pull alyanktk/welcome-app:latest
```

Then check:

```bash
docker images
```

You should see:

```text
alyanktk/welcome-app    latest
```

---

# 10. Run the Image from Docker Hub

After pulling it:

```bash
docker run -p 5000:5000 alyanktk/welcome-app:latest
```

Docker will create a container from the image.

The overall concept is:

```text
Dockerfile
    ↓
docker build
    ↓
Image
    ↓
docker tag
    ↓
Docker Hub name
    ↓
docker push
    ↓
Docker Hub
    ↓
docker pull
    ↓
Another computer
    ↓
docker run
    ↓
Container
```

# Important Commands

| Command | Purpose |
|---|---|
| `docker build` | Build an image |
| `docker images` | List local images |
| `docker login` | Login to Docker Hub |
| `docker tag` | Give an image a Docker Hub-compatible name |
| `docker push` | Upload an image to Docker Hub |
| `docker pull` | Download an image from Docker Hub |
| `docker run` | Create and run a container |
| `docker info` | Check Docker daemon information |

### Most important syntax

```bash
docker build -t <image-name>:<tag> .
```

```bash
docker tag <image-name>:<tag> <username>/<repository>:<tag>
```

```bash
docker push <username>/<repository>:<tag>
```

```bash
docker pull <username>/<repository>:<tag>
```

```bash
docker run <username>/<repository>:<tag>
```
