# 🐳 Dockerfile Commands & Workflow

A `Dockerfile` is a text file containing instructions that Docker uses to **build a Docker Image**.

The basic flow is:

```text
Dockerfile
     ↓
docker build
     ↓
Docker Image
     ↓
docker run
     ↓
Docker Container
```

---

# 📁 Example Project

Suppose our Flask project looks like this:

```text
my-flask-app/
│
├── Dockerfile
├── app.py
└── requirements.txt
```

`requirements.txt`:

```text
Flask
```

`app.py`:

```python
from flask import Flask

app = Flask(__name__)

@app.route("/")
def hello():
    return "Hello, World!"

if __name__ == "__main__":
    app.run(host="0.0.0.0", port=5000)
```

---

# 🐳 Dockerfile

Create a file named:

```text
Dockerfile
```

```dockerfile
# --------------------------------------------------
# 1. BASE IMAGE
# --------------------------------------------------

# Use Python 3.8 with Alpine Linux as the base image.
#
# python → Python Docker image
# 3.8    → Python version
# alpine → Small Linux distribution
#
FROM python:3.8-alpine


# --------------------------------------------------
# 2. COPY FILES
# --------------------------------------------------

# Copy everything from the current local directory
# into /app inside the Docker image.
#
# Syntax:
#
# COPY <source> <destination>
#
COPY . /app


# --------------------------------------------------
# 3. WORKING DIRECTORY
# --------------------------------------------------

# Set /app as the current working directory.
#
# After this:
#
# RUN commands
# CMD commands
#
# will work relative to /app.
#
WORKDIR /app


# --------------------------------------------------
# 4. INSTALL DEPENDENCIES
# --------------------------------------------------

# Install all Python packages listed in
# requirements.txt.
#
# This command runs while BUILDING the image.
#
RUN pip install -r requirements.txt


# --------------------------------------------------
# 5. DEFAULT COMMAND
# --------------------------------------------------

# Run the Flask application when the container starts.
#
# This command runs when the CONTAINER starts,
# not when the image is being built.
#
CMD ["python", "app.py"]
```

---

# 🧱 Dockerfile Instructions

The most important Dockerfile commands are:

```text
FROM
COPY
ADD
WORKDIR
RUN
CMD
ENTRYPOINT
EXPOSE
ENV
ARG
```

You don't need to memorize all of them immediately.

Start with:

```text
FROM
COPY
WORKDIR
RUN
CMD
```

---

# 1. FROM

## What is it?

Specifies the **base image** for your Docker image.

## Syntax

```dockerfile
FROM <image>
```

## Example

```dockerfile
FROM python:3.8-alpine
```

Meaning:

```text
Start with:
Python 3.8
+
Alpine Linux
```

Another example:

```dockerfile
FROM ubuntu:24.04
```

---

# 2. COPY

## What is it?

Copies files from your **local machine** into the Docker image.

## Syntax

```dockerfile
COPY <source> <destination>
```

## Example

```dockerfile
COPY . /app
```

Meaning:

```text
Local machine
     │
     │ everything in current directory
     ↓
Docker Image
     │
     ↓
/app
```

Example:

```text
Local Project

myapp/
├── app.py
├── requirements.txt
└── Dockerfile

          ↓ COPY . /app

Docker Image

/app/
├── app.py
├── requirements.txt
└── Dockerfile
```

---

# 3. WORKDIR

## What is it?

Sets the **working directory** inside the image/container.

## Syntax

```dockerfile
WORKDIR <directory>
```

## Example

```dockerfile
WORKDIR /app
```

Now `/app` becomes the current directory.

For example:

```dockerfile
WORKDIR /app

RUN pip install -r requirements.txt
```

is effectively working with:

```text
/app/requirements.txt
```

---

# 4. RUN

## What is it?

Executes a command **while the Docker image is being built**.

## Syntax

```dockerfile
RUN <command>
```

## Example

```dockerfile
RUN pip install -r requirements.txt
```

Docker executes:

```text
pip
 ↓
install
 ↓
requirements.txt
```

and installs the dependencies into the image.

---

# ⚠️ RUN vs CMD

This is extremely important.

### RUN

Runs during:

```text
docker build
```

Example:

```dockerfile
RUN pip install -r requirements.txt
```

### CMD

Runs when:

```text
docker run
```

starts a container.

Example:

```dockerfile
CMD ["python", "app.py"]
```

Think:

```text
docker build
     │
     ↓
    RUN
     │
     ↓
Docker Image
     │
     │ docker run
     ↓
    CMD
     │
     ↓
Container
```

---

# 5. CMD

## What is it?

Defines the default command that runs when a container starts.

## Syntax

```dockerfile
CMD ["command", "argument"]
```

## Example

```dockerfile
CMD ["python", "app.py"]
```

When we run:

```bash
docker run myapp
```

Docker executes:

```bash
python app.py
```

inside the container.

---

# 6. EXPOSE

## What is it?

Documents which port the application inside the container is expected to use.

## Syntax

```dockerfile
EXPOSE <port>
```

Example:

```dockerfile
EXPOSE 5000
```

If Flask runs on:

```text
5000
```

we can document it with:

```dockerfile
EXPOSE 5000
```

### Important

`EXPOSE` **does not publish the port to your computer**.

You still need:

```bash
docker run -p 5000:5000 myapp
```

---

# 7. ENV

## What is it?

Sets an environment variable inside the image/container.

## Syntax

```dockerfile
ENV VARIABLE=value
```

Example:

```dockerfile
ENV APP_ENV=production
```

Then the application can access:

```text
APP_ENV=production
```

---

# 8. ARG

## What is it?

Defines a variable that can be passed during the **image build**.

## Syntax

```dockerfile
ARG VARIABLE=value
```

Example:

```dockerfile
ARG PYTHON_VERSION=3.12
```

`ARG` is mainly available during the build process.

---

# 9. ADD

## What is it?

`ADD` can copy files into the image, similar to `COPY`.

## Syntax

```dockerfile
ADD <source> <destination>
```

Example:

```dockerfile
ADD . /app
```

For beginners:

> Prefer `COPY` unless you specifically need functionality provided by `ADD`.

# 🏗️ Building a Docker Image

Once you have:

```text
Dockerfile
app.py
requirements.txt
```

open the terminal in that directory.

Example:

```bash
cd my-flask-app
```

Then build:

```bash
docker build -t myflaskapp .
```

---

# 🔍 Breaking Down docker build

```bash
docker build -t myflaskapp .
```

Breakdown:

```text
docker
  ↓
Docker CLI

build
  ↓
Build an image

-t
  ↓
Assign a name/tag

myflaskapp
  ↓
Image name

.
  ↓
Current directory / build context
```

So:

```bash
docker build -t myflaskapp .
```

means:

> Build a Docker image using the Dockerfile in the current directory and name the image `myflaskapp`.

---

# 🏷️ Image Name and Tag

You can specify a tag:

```bash
docker build -t myflaskapp:1.0 .
```

Here:

```text
myflaskapp
     ↓
Image name

1.0
     ↓
Tag
```

Another example:

```bash
docker build -t myflaskapp:latest .
```

---

# 🖼️ Check the Image

After building:

```bash
docker images
```

You should see something similar to:

```text
REPOSITORY    TAG       IMAGE ID       SIZE
myflaskapp    latest    abc123...      60MB
```

---

# 🚀 Create and Run a Container

Once the image exists:

```bash
docker run myflaskapp
```

This:

```text
Docker Image
     │
     │ docker run
     ↓
Docker Container
```

creates a new container from the image.

---

# 🌐 Run Flask Container with Port Mapping

For our Flask application:

```bash
docker run -p 5000:5000 myflaskapp
```

Or in the background:

```bash
docker run -d -p 5000:5000 myflaskapp
```

---

# 🔌 Understanding -p

The syntax is:

```bash
-p HOST_PORT:CONTAINER_PORT
```

Example:

```bash
-p 5000:5000
```

means:

```text
Your Computer
localhost:5000
      │
      ↓
Docker
      │
      ↓
Container
port 5000
```

So you can visit:

```text
http://localhost:5000
```

---

# 🔄 Another Port Example

Suppose your Flask application runs inside the container on:

```text
5000
```

but you want to access it from your computer using:

```text
8000
```

Use:

```bash
docker run -p 8000:5000 myflaskapp
```

Meaning:

```text
Host                  Container

8000  ──────────────→  5000
```

You visit:

```text
http://localhost:8000
```

---

# 💤 Run Container in Background

## Command

```bash
docker run -d -p 5000:5000 myflaskapp
```

### `-d`

Means:

```text
detached mode
```

The container runs in the background.

Without `-d`:

```bash
docker run -p 5000:5000 myflaskapp
```

your terminal stays attached to the container.

With `-d`:

```bash
docker run -d -p 5000:5000 myflaskapp
```

you immediately get your terminal back.

```text
Don't use previous build cache.
Build the layers again.
```

# 🧠 Dockerfile Commands vs Docker CLI Commands

Don't confuse these two.

## Dockerfile Instructions

These go **inside the Dockerfile**:

```dockerfile
FROM
COPY
WORKDIR
RUN
CMD
EXPOSE
ENV
ARG
ADD
```

Example:

```dockerfile
FROM python:3.8-alpine
COPY . /app
WORKDIR /app
RUN pip install -r requirements.txt
EXPOSE 5000
CMD ["python", "app.py"]
```

---

## Docker CLI Commands

These are typed in your **terminal**:

```bash
docker build
docker run
docker ps
docker stop
docker start
docker rm
docker images
docker pull
docker push
docker logs
docker exec
docker rmi
```

# Docker – Push an Image to Docker Hub

## 1. Start Docker

Before doing anything, make sure the Docker daemon is running.

For Docker Desktop on Linux:

```bash
systemctl --user start docker-desktop
```

Check that Docker is running:

```bash
docker info
```

If you see a `Server:` section, Docker is running.

---

## 2. Build the Docker Image

First, build your image from the directory containing your `Dockerfile`.

```bash
docker build -t welcome-app:latest .
```

### Breakdown

```text
docker build
```

Build a Docker image.

```text
-t welcome-app:latest
```

Give the image a name and tag.

```text
.
```

Use the current directory as the build context.

You can check the image:

```bash
docker images
```

You should see:

```text
welcome-app    latest    ...
```

---

## 3. Login to Docker Hub

Login to your Docker Hub account:

```bash
docker login
```

Docker may give you a device code and ask you to authenticate through your browser.

After successful authentication:

```text
Login Succeeded
```

### Important

Docker needs to store your credentials locally.

On Linux, this may require a configured credential store such as `pass` + GPG.

If you get:

```text
pass not initialized
```

initialize `pass` with your GPG key:

```bash
pass init YOUR_GPG_KEY_ID
```

Then login again:

```bash
docker login
```

---

# 4. Tag the Image

Your local image might currently be:

```text
welcome-app:latest
```

Docker Hub needs the image to be tagged with your Docker Hub username.

The format is:

```bash
docker tag <local-image>:<tag> <dockerhub-username>/<repository>:<tag>
```

For example:

```bash
docker tag welcome-app:latest alyanktk/welcome-app:latest
```

Now check:

```bash
docker images
```

You should see something similar to:

```text
REPOSITORY              TAG       IMAGE ID
welcome-app             latest    5bf45c1b6e30
alyanktk/welcome-app    latest    5bf45c1b6e30
```

Both images have the **same IMAGE ID**.

The second one is simply another tag pointing to the same image.

---

# 5. Push the Image

Now push the Docker image to Docker Hub:

```bash
docker push alyanktk/welcome-app:latest
```

The general syntax is:

```bash
docker push <dockerhub-username>/<repository>:<tag>
```

Example:

```bash
docker push alyanktk/welcome-app:latest
```

Docker will upload the image layers to Docker Hub.

You may see:

```text
The push refers to repository [docker.io/alyanktk/welcome-app]

86e037ebf8b6: Pushed
4f4fb700ef54: Pushed
1da342a38765: Pushed
...
latest: digest: sha256:...
```

---

# 6. Why Do We Need to Tag the Image?

If you run:

```bash
docker push welcome-app:latest
```

Docker interprets it as:

```text
docker.io/library/welcome-app
```

The `library` namespace is generally for official Docker images.

You don't own that namespace, so you may get:

```text
push access denied
```

Instead:

```bash
docker tag welcome-app:latest alyanktk/welcome-app:latest
```

Now Docker knows:

```text
Docker Hub
    │
    └── alyanktk
          │
          └── welcome-app
```

So:

```bash
docker push alyanktk/welcome-app:latest
```

pushes to your Docker Hub namespace.

---

# 7. Complete Workflow

From beginning to end:

```bash
# Start Docker Desktop
systemctl --user start docker-desktop

# Check Docker
docker info

# Build the image
docker build -t welcome-app:latest .

# Check images
docker images

# Login to Docker Hub
docker login

# Tag the image
docker tag welcome-app:latest alyanktk/welcome-app:latest

# Check the tag
docker images

# Push to Docker Hub
docker push alyanktk/welcome-app:latest
```

---

# 8. Understanding the Whole Process

The process is:

```text
                Local Computer
                     │
                     │
                Dockerfile
                     │
                     ↓
              docker build
                     │
                     ↓
             Docker Image
          welcome-app:latest
                     │
                     │ docker tag
                     ↓
       alyanktk/welcome-app:latest
                     │
                     │ docker push
                     ↓
                Docker Hub
                     │
                     ↓
       alyanktk/welcome-app
```

---

# 9. Pull the Image Later

Once the image is on Docker Hub, another computer can download it using:

```bash
docker pull alyanktk/welcome-app:latest
```

Then check:

```bash
docker images
```

You should see:

```text
alyanktk/welcome-app    latest
```

---

# 10. Run the Image from Docker Hub

After pulling it:

```bash
docker run -p 5000:5000 alyanktk/welcome-app:latest
```

Docker will create a container from the image.

The overall concept is:

```text
Dockerfile
    ↓
docker build
    ↓
Image
    ↓
docker tag
    ↓
Docker Hub name
    ↓
docker push
    ↓
Docker Hub
    ↓
docker pull
    ↓
Another computer
    ↓
docker run
    ↓
Container
```

| `docker build` | Build an image |

```bash
docker build -t <image-name>:<tag> .
```

```bash
docker tag <image-name>:<tag> <username>/<repository>:<tag>
```

```bash
docker push <username>/<repository>:<tag>
```
