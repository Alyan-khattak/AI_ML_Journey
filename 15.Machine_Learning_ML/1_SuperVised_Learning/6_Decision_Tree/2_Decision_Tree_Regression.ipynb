# Decision Tree Regression

Decision Tree Regression is a **Supervised Machine Learning Regression Algorithm**.

It is used when the target variable is **continuous (numerical)**.

Examples:

- House Price Prediction
- Salary Prediction
- Temperature Prediction
- Stock Price Estimation
- Sales Forecasting

Unlike Decision Tree Classifier, it predicts a **number**, not a class.

---

# Intuition

Suppose we want to predict **House Price**.

| House Size (sqft) | Price ($) |
|-------------------|-----------|
| 800 | 120,000 |
| 1000 | 150,000 |
| 1200 | 180,000 |
| 1800 | 300,000 |
| 2200 | 370,000 |

The tree keeps asking questions like:

```
House Size ≤ 1500?

        Yes                No
         │                  │
Average = $150,000    Average = $335,000
```

Instead of predicting a class,

it predicts the **average value of the samples in the leaf node**.

---

# How Does It Work?

```
Dataset
    │
    ▼
Find Best Split
    │
    ▼
Split Dataset
    │
    ▼
Repeat Until Stopping Condition
    │
    ▼
Leaf Node
    │
    ▼
Predict Average Target Value
```

---

# How Does It Choose the Best Split?

Decision Tree Regression does **NOT** use

- Gini Impurity ❌
- Entropy ❌
- Information Gain ❌

Instead,

it uses **Variance Reduction**.

---

# What is Variance?

Variance measures **how spread out the target values are**.

If all values are close together,

↓

Low Variance.

If values are very different,

↓

High Variance.

Example

Low Variance

```
100

102

101

99
```

High Variance

```
50

200

400

900
```

---

# Goal of Decision Tree Regression

The tree tries to create child nodes where the target values are as similar as possible.

```
Mixed Values

↓

Split

↓

More Similar Values
```

---

# Variance Reduction

A good split should **reduce the variance**.

Formula

$$
\text{Variance Reduction}
=
\text{Parent Variance}
-
\text{Weighted Child Variance}
$$

The split with the **highest Variance Reduction** is chosen.

---

# Simple Example

Suppose we have house prices

```
100

110

120

520

```

Parent Node

```
High Variance
```

```

```

```

```


↓

```

```

Therefore,

this is a **good split**.

# Hyperparameters

## max_depth

Maximum depth of the tree.

```python
max_depth=5
```

---

## min_samples_split

Minimum samples required before splitting.

```python
min_samples_split=5
```

---

## min_samples_leaf

Minimum samples required in each leaf.

```python
min_samples_leaf=2
```

---

## max_leaf_nodes

Maximum number of leaf nodes.

```python
max_leaf_nodes=20
```

---

## random_state

Ensures reproducible results.

```python
random_state=42
```

---

# Scikit-Learn Syntax

```python
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor(
    max_depth=5,
    random_state=42
)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

---

# Hyperparameter Tuning

```python
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeRegressor

params = {
    "max_depth": [3, 5, 7, 10],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 5]
}

grid = GridSearchCV(
    DecisionTreeRegressor(random_state=42),
    params,
    cv=5,
    scoring="neg_mean_squared_error"
)

grid.fit(X_train, y_train)

print(grid.best_params_)
```

---

# Advantages

✅ Easy to understand

✅ Handles non-linear relationships

✅ No feature scaling required

✅ Works with numerical and categorical features

---

# Disadvantages

❌ Easily overfits

❌ Sensitive to noisy data

❌ Produces step-like predictions instead of smooth curves

---

# Decision Tree Classifier vs Regressor

| Decision Tree Classifier | Decision Tree Regressor |
|--------------------------|-------------------------|
| Predicts a class | Predicts a number |
| Uses Gini or Entropy | Uses Variance Reduction |
| Output = Category | Output = Continuous Value |

---

# One-Line Definition

> **Decision Tree Regression is a supervised learning algorithm that predicts continuous values by recursively splitting the data into groups that minimize variance, with each leaf node predicting the average target value of the samples it contains.**
