# 🐳 Docker Image

## What is a Docker Image?

A **Docker Image** is a read-only template/blueprint used to create Docker containers.

It contains everything required for an application to run, such as:

- Base operating system files
- Runtime
- Libraries
- Dependencies
- Application code
- Configuration

In simple words:

> **Docker Image = Everything needed to create and run an application inside a container.**

```text
Docker Image
     │
     ├── Operating System files
     ├── Python
     ├── Libraries
     ├── Dependencies
     ├── Application Code
     └── Configuration
```

---

# 🧱 Docker Image Layers

One of the most important concepts about Docker Images is that they are built using **layers**.

A Docker Image is not just one big block.

It is made up of multiple layers stacked on top of each other.

```text
┌──────────────────────────────┐
│      Application Code        │  ← Layer
├──────────────────────────────┤
│      Python Dependencies     │  ← Layer
├──────────────────────────────┤
│      Python Runtime          │  ← Layer
├──────────────────────────────┤
│      Base Image / OS Files   │  ← Layer
└──────────────────────────────┘
```

All these layers together form the:

```text
        🐳 Docker Image
```

---

# 🧩 What is an Image Layer?

An **image layer** is a change or set of files added to an image during the image-building process.

For example, imagine we start with:

```text
Ubuntu
```

Then install Python:

```text
Ubuntu
   ↓
Python added
```

Then install dependencies:

```text
Ubuntu
   ↓
Python
   ↓
Pandas
   ↓
NumPy
```

Then add our application:

```text
Ubuntu
   ↓
Python
   ↓
Pandas + NumPy
   ↓
Application Code
```

Each step can create a new layer.

# 🏗️ Example Dockerfile

Consider this Dockerfile:

```dockerfile
FROM python:3.12

WORKDIR /app

COPY requirements.txt .

RUN pip install -r requirements.txt

COPY . .

CMD ["python", "app.py"]
```

Docker uses these instructions to build the image.

Conceptually:

```text
Dockerfile
     │
     ├── FROM python:3.12
     │        ↓
     │     Base Layer
     │
     ├── WORKDIR /app
     │        ↓
     │     Layer
     │
     ├── COPY requirements.txt .
     │        ↓
     │     Layer
     │
     ├── RUN pip install -r requirements.txt
     │        ↓
     │     Dependencies Layer
     │
     ├── COPY . .
     │        ↓
     │     Application Layer
     │
     └── CMD ["python", "app.py"]
              ↓
          Configuration
```

The final result is one Docker Image made from these layers.

# ♻️ Why Does Docker Use Layers?

Layers provide several important benefits.

## 1. Reusability

Docker can reuse layers that already exist.

For example:

```text
Image A
├── Python Layer
├── Pandas Layer
└── App A Layer
```

and:

```text
Image B
├── Python Layer
├── Pandas Layer
└── App B Layer
```

The Python and Pandas layers can potentially be reused.

```text
          Python Layer
               │
       ┌───────┴────────┐
       ↓                ↓
    Image A           Image B
```

This avoids storing duplicate data unnecessarily.

---

# 2. Faster Builds

Suppose you build an image:

```text
Base Image
    ↓
Python
    ↓
Dependencies
    ↓
Application
```

Later you only change:

```text
Application Code
```

Docker can reuse the unchanged layers instead of rebuilding everything from scratch.

```text
OLD BUILD

Base       ✅ Reuse
Python     ✅ Reuse
Packages   ✅ Reuse
App        🔨 Rebuild
```

This is one reason Docker builds can become much faster when Docker's **build cache** is used effectively.

---

# 3. Storage Efficiency

If multiple images use the same layer, Docker can store that shared layer once and reuse it.

```text
              Python Layer
                   │
        ┌──────────┼──────────┐
        ↓          ↓          ↓
     Image A    Image B    Image C
```

Instead of:

```text
Image A → Python copy
Image B → Python copy
Image C → Python copy
```

Docker can share the common layer.

---

# 🔒 Layers Are Immutable

Docker image layers are generally treated as **immutable**.

This means that once a layer has been created, Docker doesn't normally modify that existing layer.

If something changes, a new layer can be created.

Conceptually:

```text
Layer 1
   ↓
Layer 2
   ↓
Layer 3
```

If you change something in Layer 3:

```text
Layer 1   ← unchanged
   ↓
Layer 2   ← unchanged
   ↓
New Layer 3
```

---

# 📝 Dockerfile → Layers → Image

The complete process can be understood as:

```text
Dockerfile
    │
    │
    ▼
Docker Build
    │
    ├── Layer 1
    ├── Layer 2
    ├── Layer 3
    ├── Layer 4
    └── Layer 5
          │
          ▼
     Docker Image
```

Then:

```text
Docker Image
      │
      │ docker run
      ▼
Docker Container
      │
      ▼
Running Application
```

---

# 🆚 Image vs Container

Remember:

```text
Docker Image
     │
     │ docker run
     ▼
Docker Container
```

### Image

```text
Blueprint / Template
       ↓
Read-only layers
       ↓
Used to create containers
```

### Container

```text
Running instance of an image
       ↓
Uses the image's layers
       ↓
Has a writable container layer
```

Conceptually:

```text
          Container
┌──────────────────────────────┐
│ Writable Container Layer     │
├──────────────────────────────┤
│ Image Layer                  │
├──────────────────────────────┤
│ Image Layer                  │
├──────────────────────────────┤
│ Image Layer                  │
├──────────────────────────────┤
│ Base Image Layer             │
└──────────────────────────────┘
```

The image layers remain unchanged while the running container can have its own writable layer on top.

## One-Line Definition

> **A Docker Image is a read-only, layered template containing the files, runtime, dependencies, and configuration needed to create and run a container.**

## The Key Idea

```text
             🐳 DOCKER IMAGE
                    │
       ┌────────────┼────────────┐
       ↓            ↓            ↓
    Layer 1      Layer 2      Layer 3
       │            │            │
       └────────────┼────────────┘
                    ↓
              One Image
                    ↓
             Docker Container
```

> **Many layers combine to form one Docker Image, and a Docker Container is created from that image.**
