# WordNet Lemmatization

## 1. What is Lemmatization?

**Lemmatization** reduces a word to its **base/dictionary form**, called a **lemma**.

Unlike stemming, lemmatization tries to return a **real English word**.

> We will Get meaningfull word ( the Output is called **Lemma**)

### Example

```text
Stemming:

studies → studi
```

```text
Lemmatization:

studies → study
```

Another example:

```text
running → running
```

Depending on the context/POS, it can become:

```text
running → run
```

---

# 2. WordNet

**WordNet** is a large lexical database of English words.

NLTK uses WordNet to help perform lemmatization.

```text
Word
 ↓
WordNet
 ↓
Find base/dictionary form
 ↓
Lemma
```

# 6. More Examples

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = [
    "cars",
    "dogs",
    "studies",
    "running",
    "better"
]

for word in words:
    print(word, "→", lemmatizer.lemmatize(word))
```

The result may not always be what you expect:

```text
cars → car
dogs → dog
studies → study
running → running
better → better
```

Why?

Because **WordNet needs to know the Part of Speech (POS)** to correctly understand the word.

---

# 7. POS — Part of Speech

POS tells us what type of word something is.

Common POS:

```text
n → Noun
v → Verb
a → Adjective
r → Adverb
```

Example:

```python
lemmatizer.lemmatize("running", pos="v")
```

Output:

```text
run
```

Without POS:

```python
lemmatizer.lemmatize("running")
```

It assumes:

```text
noun
```

So it may return:

```text
running
```

---

# 8. Using POS

### Noun

```python
lemmatizer.lemmatize("cars", pos="n")
```

Output:

```text
car
```

### Verb

```python
lemmatizer.lemmatize("running", pos="v")
```

Output:

```text
run
```

### Adjective

```python
lemmatizer.lemmatize("better", pos="a")
```

Output:

```text
good
```

### Adverb

```python
lemmatizer.lemmatize("quickly", pos="r")
```

Output:

```text
quickly
```

---

# 9. Important Difference

## Stemming

```text
studies
   ↓
studi
```

It simply cuts/changes word endings using rules.

## Lemmatization

```text
studies
   ↓
study
```

It tries to find the correct dictionary/base form.

---

# 10. Stemming vs Lemmatization

| Stemming | Lemmatization |
|---|---|
| Faster | Usually slower |
| Uses rules | Uses linguistic/dictionary information |
| Can produce invalid words | Usually produces real words |
| Less accurate | More accurate |
| No POS usually required | POS can be important |

---

# 11. Simple Example

```python
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

word = "studies"

print("Stemming:", stemmer.stem(word))
print("Lemmatization:", lemmatizer.lemmatize(word))
```

Possible output:

```text
Stemming: studi
Lemmatization: study
```

# 12. Lemmatizing Multiple Words


---

# 13. Lemmatization with POS

```python
words = [
    ("running", "v"),
    ("studies", "v"),
    ("better", "a"),
    ("cars", "n")
]

for word, pos in words:
    print(
        word,
        "→",
        lemmatizer.lemmatize(word, pos=pos)
    )
```

Possible output:

```text
running → run
studies → study
better → good
cars → car
```
