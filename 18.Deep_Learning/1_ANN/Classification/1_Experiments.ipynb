# Sequential and Dense in Keras

## Sequential

`Sequential` is used to **build the neural network by adding layers one after another**.

```text
Input
  ↓
Dense Layer
  ↓
Dense Layer
  ↓
Output
```

### SynTax:

```python 
model = Sequential([
    layer1,
    layer2,
    layer3
])
```


## Dense

Dense represents a fully connected layer of neurons.

Each neuron receives input from all neurons in the previous layer.

### Syntax

```python 
Dense(
    number_of_neurons,
    activation="activation_function"
)

```

## Using Both Together

> We use Sequential to create the model and Dense to create its layers.

```Text
10 Input Features
       ↓
Dense → 64 Neurons → ReLU
       ↓
Dense → 32 Neurons → ReLU
       ↓
Dense → 1 Neuron → Sigmoid
       ↓
Prediction

```

> Sequential = the structure/container of the ANN

> Dense = the actual layers/neurons inside the ANN

___

````python
# Build ANN model 

model = Sequential(
    [
    Dense(64, activation="relu", input_shape=(X_train_scaled.shape[1],)), #HL1 Connected with INput Layer
    Dense(32, activation="relu"), # HL2 with 32 nodes
    Dense(1,activation="sigmoid") # Output Layer
    ]
)
````

# Model Compilation

`compile()` tells the ANN **how to learn**.

It mainly defines:

- **Optimizer** → How weights are updated
- **Loss** → How wrong the predictions are
- **Metrics** → How performance is measured

### Syntax

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

```
```
To do forward and Backword Propagation we need to Compile the Model
```

> Remember: compile() prepares the model for training; it does not train the model

___

````python
# To Indrectly Set and Optimizer, Activatvation Functions and Loss Function we can use this 


import tensorflow
opt = tensorflow.keras.optimizers.Adam(learning_rate = 0.01)
loss = tensorflow.keras.losses.binary_crossentropy()
````

````python
# to Direclty set it we can use this 

model.compile(
    optimizer=opt,
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
````

## Setup Tensor Board Logs before Training


````python
import datetime

from tensorflow.keras.callbacks import EarlyStopping, TensorBoard


# Create a unique directory for TensorBoard logs
# Example:
# logs/fit/20260903-151200
log_dir = "logs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")


# Create the TensorBoard callback
# histogram_freq=1 means TensorBoard will record
# weight/activation histograms after every epoch.
tensorflow_callback = TensorBoard(
    log_dir=log_dir,
    histogram_freq=1
)
````

# Early Stopping

## What is Early Stopping?

**Early Stopping** is a technique used during model training to **stop training automatically when the model stops improving**.

Instead of training for a fixed number of epochs, we monitor a metric such as validation loss.

---

## Why is it Used?

It is mainly used to:

- Prevent **overfitting**
- Stop unnecessary training
- Save training time
- Keep the model at its **best performance**

---

## How It Works

```text
Epoch 1 → Validation Loss ↓
Epoch 2 → Validation Loss ↓
Epoch 3 → Validation Loss ↓
Epoch 4 → Validation Loss ↓
Epoch 5 → Validation Loss ↑
Epoch 6 → Validation Loss ↑
Epoch 7 → Validation Loss ↑
                 ↓
          Stop Training
```

> Usually, we use patience to allow the model a few epochs to improve before stopping.


# Important Parameters

### monitor

> The metric we want to watch.

```text
val_loss
val_accuracy
loss
accuracy
```

### patience

> Number of epochs to wait for improvement before stopping.

``` python
patience=5

If the monitored metric doesn't improve for 5 consecutive epochs, stop training.

```

### restore_best_weights

> Restores the model weights from the epoch where the monitored metric was best

## Basic Synatx

```python 
from tensorflow.keras.callbacks import EarlyStopping

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

```

> Early Stopping prevents unnecessary training and helps reduce overfitting by stopping when validation performance stops improving.

````python
# Implement Early Stoping 

early_stopping_callback = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)
````

# Training The Model 


## Model Fit

`fit()` is used to **train the ANN on the training data**.

### Syntax

```python
model.fit(
    X_train,
    y_train,
    epochs=100,
    validation_data=(X_test, y_test)
)

X_train → Training features
y_train → Training labels
epochs → Number of times the model sees the complete training dataset
validation_data → Data used to check performance after each epoch
```

> Remember: compile() tells the model how to learn, while fit() actually trains the model.

___
___

````python

history = model.fit(
    X_train_scaled,
    y_train,
    validation_data=(X_test_scaled, y_test),
    epochs=100,
    callbacks=[tensorflow_callback, early_stopping_callback]
)
````
