# Word2Vec

## 1. What is Word2Vec?

**Word2Vec** is a technique used in NLP to represent words as **numerical vectors**.

Instead of representing a word as:

```text
"king"
```

Word2Vec represents it as a vector:

```text
king → [0.21, -0.45, 0.73, 0.11, ...]
```

These vectors are called **word embeddings**.

The important idea is:

> Words that have similar meanings or occur in similar contexts tend to have similar vector representations.

For example:

```text
king
queen
prince
```

will generally have more similar vectors than:

```text
king
banana
```

---

# 2. Why Do We Need Word2Vec?

Traditional text representations such as **Bag of Words** and **TF-IDF** represent words using individual features.

For example:

```text
Vocabulary:

["king", "queen", "apple", "banana"]
```

A simple representation might be:

```text
king   → [1, 0, 0, 0]
queen  → [0, 1, 0, 0]
apple  → [0, 0, 1, 0]
banana → [0, 0, 0, 1]
```

The problem is that these vectors don't tell us that:

```text
king
queen
```

are semantically related.

They are simply different positions in the vocabulary.

Word2Vec tries to learn **meaning from context**.

---

# 3. What is Feature Representation?

A **feature representation** means converting something into numerical values that a Machine Learning model can work with.

For text:

```text
Word
 ↓
Numerical representation
 ↓
Feature vector
```

For example:

```text
"king"
```

could be represented as:

```text
[0.21, -0.45, 0.73, 0.11, -0.62]
```

Each number is a learned feature.

These features don't normally have simple human-readable names such as:

```text
feature 1 = royalty
feature 2 = gender
feature 3 = power
```

Instead, Word2Vec learns numerical dimensions that collectively capture patterns in word usage.

---

# 4. Word Embedding

The vector representation of a word is called its **word embedding**.

Example:

```text
king → [0.21, -0.45, 0.73, 0.11]
queen → [0.19, -0.42, 0.70, 0.14]
apple → [-0.81, 0.22, -0.14, 0.65]
```

The actual values are learned from the training corpus.

The dimensions are not manually assigned.

---

# 5. Main Intuition

Word2Vec follows the idea:

> **You can learn something about a word by looking at the words that appear around it.**

Consider:

```text
The king ruled the kingdom.
The king wore a crown.
The queen ruled the kingdom.
The queen wore a crown.
```

The words:

```text
king
queen
```

appear in similar contexts.

Therefore Word2Vec learns vectors that are relatively close to each other.

---

# 6. Context

The **context** of a word means the words surrounding it.

Consider:

```text
The cat is sitting on the mat.
```

For:

```text
cat
```

the nearby words might be:

```text
The
is
sitting
```

Depending on the selected window size.

Word2Vec uses these relationships to learn word representations.

---

# 7. Window Size

**Window size** determines how many nearby words Word2Vec considers around a target word.

Example:

```text
I love machine learning very much
```

Suppose:

```text
window_size = 2
```

For the target:

```text
machine
```

the context can include:

```text
love
learning
```

and potentially words farther away according to the selected window.

A simple visualization:

```text
I     love     machine     learning     very     much
      ←────       ↑        ────→
             context
```

---

# 8. Example of Window Size

Sentence:

```text
I love machine learning
```

Suppose:

```text
window_size = 1
```

For:

```text
machine
```

the nearby context is:

```text
love
learning
```

For:

```text
learning
```

the nearby context is:

```text
machine
```

Increasing the window allows Word2Vec to consider more distant words.

---

# 9. Why Window Size Matters

Small window:

```text
window = 2
```

Usually focuses more on **local/syntactic relationships**.

Larger window:

```text
window = 5
```

can capture more **semantic/topic-level relationships**.

There isn't one universally best window size.

It depends on the dataset and task.

---

# 10. How Word2Vec Learns

Word2Vec learns embeddings by using a prediction task.

The basic idea is:

```text
Words in context
       ↓
Prediction task
       ↓
Adjust model weights
       ↓
Learn useful word vectors
```

There are two major Word2Vec architectures:

```text
Word2Vec
   │
   ├── CBOW
   │
   └── Skip-Gram
```

---

# 11. CBOW

**CBOW = Continuous Bag of Words**

CBOW uses the **surrounding context words to predict the target word**.

In simple terms:

```text
Context words
      ↓
Predict missing/target word
```

---

# 12. CBOW Example

Consider:

```text
I love machine learning
```

Suppose the target word is:

```text
machine
```

and the context is:

```text
I
love
learning
```

The model tries to learn:

```text
I + love + learning
          ↓
       machine
```

So:

```text
Context → Target
```

---

# 13. CBOW Training Example

Suppose:

```text
The cat is sitting on the mat
```

Using a suitable window, we can create training examples such as:

```text
Context                  Target

The, is                  cat
cat, sitting             is
is, on                   sitting
sitting, the             on
on, mat                  the
```

The exact training pairs depend on the window size and implementation.

The model learns to predict the target word from its surrounding words.

---

# 14. CBOW Intuition

Think:

```text
"I am ___ machine learning."

What word should go here?
```

The model uses the surrounding words to predict:

```text
learning
```

or whichever word is selected as the target.

During training, the model repeatedly performs these prediction tasks.

The learned weights become useful word representations.

---

# 15. Skip-Gram

**Skip-Gram** works in the opposite direction.

Instead of:

```text
Context → Target
```

it uses:

```text
Target → Context
```

The target word is used to predict surrounding words.

---

# 16. Skip-Gram Example

Sentence:

```text
I love machine learning
```

Suppose:

```text
Target = machine
```

Context words might be:

```text
love
learning
```

The model tries to learn:

```text
machine → love
machine → learning
```

So:

```text
Target → Context
```

---

# 17. CBOW vs Skip-Gram

The easiest way to remember them:

```text
CBOW:

Context → Target
```

```text
Skip-Gram:

Target → Context
```

### Example

Sentence:

```text
I love machine learning
```

For `machine`:

### CBOW

```text
love + learning
       ↓
    machine
```

### Skip-Gram

```text
machine
   ↓
love
learning
```

---

# 18. CBOW vs Skip-Gram Table

| Feature        | CBOW                     | Skip-Gram                      |
| -------------- | ------------------------ | ------------------------------ |
| Full name      | Continuous Bag of Words  | Skip-Gram                      |
| Input          | Context words            | Target word                    |
| Predicts       | Target word              | Context words                  |
| Direction      | Context → Target         | Target → Context               |
| Training       | Generally faster         | Generally slower               |
| Common words   | Works well               | Works well                     |
| Rare words     | Generally less effective | Often better                   |
| Small datasets | Often useful             | Can work well                  |
| Large datasets | Efficient                | More computationally expensive |


# 21. What Actually Gets Learned?

A simplified Word2Vec model contains numerical weights.

During training:

```text
Input word
    ↓
Neural network
    ↓
Prediction
    ↓
Compare with actual context
    ↓
Calculate error
    ↓
Update weights
    ↓
Repeat
```

After training, the learned weights can be used as word embeddings.

For example:

```text
king → [0.21, -0.45, 0.73, ...]
queen → [0.19, -0.42, 0.70, ...]
```

# 23. What Does Vector Size Mean?

A larger vector provides more dimensions in which the model can encode patterns.

For example:

```text
vector_size = 50
```

versus:

```text
vector_size = 300
```

Larger isn't automatically better.

It can mean:

```text
More information
+
More computation
+
More memory
```

The appropriate size depends on the amount of training data and task.

# 25. Cosine Similarity

Word vectors are commonly compared using **cosine similarity**.

The idea is to measure the angle between two vectors.

Conceptually:

```text
Similar vectors
      ↓
Small angle
      ↓
High cosine similarity
```

and:

```text
Different vectors
      ↓
Larger angle
      ↓
Lower cosine similarity
```

# 30. Average Word2Vec

Sometimes a sentence contains multiple words:

```text
"I love machine learning"
```

But a Machine Learning model may need **one fixed-size vector representing the entire sentence**.

Word2Vec gives us:

```text
I       → vector
love    → vector
machine → vector
learning→ vector
```

One simple approach is to **average the vectors**.

This is called:

> **Average Word2Vec**

---

# 31. Average Word2Vec Intuition

Suppose:

```text
I      → [1, 2, 3]
love   → [2, 4, 6]
Python → [3, 6, 9]
```

Average them dimension by dimension:

```text
Dimension 1:

(1 + 2 + 3) / 3
= 2
```

```text
Dimension 2:

(2 + 4 + 6) / 3
= 4
```

```text
Dimension 3:

(3 + 6 + 9) / 3
= 6
```

Therefore:

```text
Sentence vector:

[2, 4, 6]
```

---

# 32. Average Word2Vec Formula

If a sentence contains `n` word vectors:

```text
v₁, v₂, v₃, ..., vₙ
```

then:

```text
Sentence Vector =
(v₁ + v₂ + v₃ + ... + vₙ) / n
```

The result has the **same number of dimensions** as the original Word2Vec vectors.

For example:

```text
Word vectors = 100 dimensions

        ↓ average

Sentence vector = 100 dimensions
```

---

# 33. Average Word2Vec Code

Suppose:

```python
sentence = ["i", "love", "machine", "learning"]
```

We can collect the vectors:

```python
vectors = [
    model.wv[word]
    for word in sentence
    if word in model.wv
]
```

Then average:

```python
import numpy as np

sentence_vector = np.mean(vectors, axis=0)
```

Check:

```python
print(sentence_vector.shape)
```

```python
vector_size=100
```

then:

```text
(100,)
```

# 34. Why Average Word2Vec?

Suppose you have:

```text
10,000 sentences
```

and want to train a classifier.

You could convert every sentence into a fixed-size vector:

```text
Sentence 1 → [100 numbers]
Sentence 2 → [100 numbers]
Sentence 3 → [100 numbers]
...
Sentence 10000 → [100 numbers]
```

Then use these vectors as features:

```text
Average Word2Vec
       ↓
Feature Matrix
       ↓
Machine Learning Model
```

For example:

```text
Average Word2Vec
       ↓
Logistic Regression
```

or:

```text
Average Word2Vec
       ↓
Random Forest
```

---

# 35. Problem With Simple Averaging

Average Word2Vec loses **word order**.

Consider:

```text
"I love Python"
```

and:

```text
"Python love I"
```

The same word vectors are averaged.

Therefore their sentence vectors can be identical or very similar.

So Average Word2Vec doesn't properly capture:

```text
word order
grammar
sentence structure
```

---

# 36. Another Problem

Consider:

```text
"I like this movie"
```

and:

```text
"I hate this movie"
```

Most words are shared:

```text
I
this
movie
```

Only:

```text
like
hate
```

changes.

Simple averaging may not represent the difference as strongly as a contextual model would.

# 37. Word2Vec vs TF-IDF

| Feature                   | TF-IDF            | Word2Vec                           |
| ------------------------- | ----------------- | ---------------------------------- |
| Representation            | Sparse            | Dense                              |
| Learns word relationships | Limited           | Yes                                |
| Semantic similarity       | Limited           | Stronger                           |
| Word order                | No                | No                                 |
| Context                   | Limited           | Uses local context during training |
| Vector size               | Vocabulary-sized  | Fixed chosen size                  |
| Handles synonyms          | Not inherently    | Better                             |
| Feature values            | Importance scores | Learned embeddings                 |
| Main idea                 | Word importance   | Word meaning/relationships         |

---

# 38. Sparse vs Dense Representation

TF-IDF might produce:

```text
[0, 0, 0, 0.72, 0, 0, 0.13, 0, 0, ...]
```

Many values are zero.

This is a **sparse representation**.

Word2Vec might produce:

```text
[0.21, -0.45, 0.73, 0.11, -0.62, ...]
```

Most values are non-zero.

This is a **dense representation**.
