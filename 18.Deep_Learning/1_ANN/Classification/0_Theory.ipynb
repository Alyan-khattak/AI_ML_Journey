# Artificial Neural Network (ANN) — Complete Notes 🧠

## 1. What is a Perceptron?

A **Perceptron** is the simplest form of an artificial neuron.

It takes:

- Input values
- Weights
- Bias

and produces an output.

The perceptron performs two main steps:

```text
Inputs
  ↓
Weighted Sum
  ↓
z = wx + b
  ↓
Activation Function
  ↓
Output
```

---

# 2. The Two Main Functions of a Perceptron

A perceptron basically performs two operations.

## Step 1 — Weighted Sum

The first function is:

```text
z = wx + b
```

Where:

```text
x = input
w = weight
b = bias
z = weighted sum
```

For multiple inputs:

```text
z = w₁x₁ + w₂x₂ + w₃x₃ + b
```

In vector form:

```text
z = WᵀX + b
```

---

## Step 2 — Activation Function

After calculating `z`, we pass it through an activation function:

```text
activation(z)
```

So:

```text
z = wx + b

output = activation(z)
```

The activation function determines the final output of the neuron.

---

# 3. Perceptron Visual

```text
                 x₁
                  │
                 w₁
                  ↓
                 ┌───┐
        x₂ ──w₂─→│ Σ │──→ z = Σ(wx) + b
                 └───┘
                  │
                  ↓
            Activation(z)
                  │
                  ↓
                Output
```

For example:

```text
x₁ ──× w₁──┐
           │
x₂ ──× w₂──┼──→ Σ + b ──→ Activation ──→ Output
           │
x₃ ──× w₃──┘
```

---

# 4. Why Do We Need Weights?

Weights determine the **importance of each input**.

For example:

```text
x₁ = study hours
x₂ = attendance
```

Maybe study hours are more important than attendance.

Therefore:

```text
w₁ = 0.8
w₂ = 0.3
```

The larger weight means that the corresponding input has a greater influence on the output.

---

# 5. Why Do We Need Bias?

Bias allows the neuron to shift the output.

Without bias:

```text
z = wx
```

With bias:

```text
z = wx + b
```

Bias gives the neuron more flexibility to learn patterns.

Think of bias as an adjustable offset.

---

# 6. Example of a Perceptron

Suppose:

```text
x₁ = 2
x₂ = 3

w₁ = 0.5
w₂ = 0.2

b = 1
```

Calculate:

```text
z = w₁x₁ + w₂x₂ + b

z = (0.5)(2) + (0.2)(3) + 1

z = 1 + 0.6 + 1

z = 2.6
```

Now:

```text
output = activation(2.6)
```

The activation function decides the final output.

---

# 7. What is an Artificial Neural Network?

An **Artificial Neural Network (ANN)** is a collection of interconnected artificial neurons.

Instead of having only one perceptron:

```text
Input → Perceptron → Output
```

we have many neurons organized into layers:

```text
Input Layer
     ↓
Hidden Layer
     ↓
Hidden Layer
     ↓
Output Layer
```

---

# 8. ANN Architecture

A typical ANN looks like:

```text
             INPUT             HIDDEN              OUTPUT

           x₁ ───────────────→ ○
                                ╲
           x₂ ───────────────→ ○ ───────────────→ ○
                                ╱
           x₃ ───────────────→ ○
```

A more complete example:

```text
Input Layer          Hidden Layer 1        Hidden Layer 2       Output

   x₁  ───────────────→ ○ ───────────────→ ○
                       ╱ ╲                 ╱ ╲
   x₂  ───────────────→ ○ ───────────────→ ○ ───────────────→ ŷ
                       ╲ ╱                 ╲ ╱
   x₃  ───────────────→ ○ ───────────────→ ○
```

---

# 9. Input Layer

The input layer receives the features.

For example, for house-price prediction:

```text
x₁ = house size
x₂ = number of bedrooms
x₃ = house age
```

These become the input neurons.

```text
Input Layer

x₁
x₂
x₃
```

---

# 10. Hidden Layer

Hidden layers perform the actual transformation and learning.

Each neuron:

```text
Weighted Sum
     ↓
Activation
     ↓
Output
```

An ANN can have:

```text
1 hidden layer
```

or:

```text
10 hidden layers
```

or many more.

When a neural network has many layers, it is generally called **Deep Learning**.

---

# 11. Output Layer

The output layer produces the prediction.

For binary classification:

```text
Output = probability
```

Example:

```text
0.87
```

Meaning:

```text
87% probability of class 1
```

For multiclass classification:

```text
Class 1 → 0.10
Class 2 → 0.75
Class 3 → 0.15
```

The highest probability is selected.

---

# 12. Forward Propagation

**Forward propagation** means passing the input from the input layer toward the output layer.

```text
Input
  ↓
Hidden Layer 1
  ↓
Hidden Layer 2
  ↓
Output
```

At every neuron:

```text
z = Wx + b
```

then:

```text
a = activation(z)
```

where:

```text
a = activation/output of neuron
```

So one layer does:

```text
Input
  ↓
Wx + b
  ↓
Activation
  ↓
Output
```

---

# 13. Forward Propagation Example

Suppose:

```text
Input → Hidden Layer → Output
```

The hidden layer calculates:

```text
z₁ = w₁x₁ + w₂x₂ + b
```

Then:

```text
a₁ = activation(z₁)
```

The output neuron receives `a₁`:

```text
z₂ = w₃a₁ + b₂
```

Then:

```text
ŷ = activation(z₂)
```

Therefore:

```text
X
 ↓
Z₁ = W₁X + b₁
 ↓
A₁ = activation(Z₁)
 ↓
Z₂ = W₂A₁ + b₂
 ↓
ŷ = activation(Z₂)
```

---

# 14. Loss / Cost Function

After forward propagation, the model produces a prediction:

```text
ŷ = predicted value
```

We compare it with the actual value:

```text
y = actual value
```

The **loss function** tells us how wrong the prediction is.

```text
Actual y
   │
   │ compare
   ↓
Prediction ŷ
   │
   ↓
Loss
```

Example:

```text
Actual = 1
Prediction = 0.9
```

Small error.

But:

```text
Actual = 1
Prediction = 0.1
```

Large error.

---

# 15. Loss vs Cost Function

### Loss

Loss usually refers to the error for **one training example**.

```text
Loss = error for one example
```

### Cost

Cost is commonly the average loss over the training examples.

```text
Cost = average of losses
```

Conceptually:

```text
Example 1 → Loss
Example 2 → Loss
Example 3 → Loss
Example 4 → Loss
       ↓
Average
       ↓
Cost
```

---

# 16. Common Loss Functions

## Regression

A common choice is:

```text
Mean Squared Error (MSE)
```

```text
MSE = (1/n) Σ(y - ŷ)²
```

---

## Binary Classification

A common choice is:

```text
Binary Cross Entropy
```

```text
Loss = -[y log(ŷ) + (1-y)log(1-ŷ)]
```

---

## Multiclass Classification

Usually:

```text
Categorical Cross Entropy
```

combined with:

```text
Softmax
```

---

# 17. Backpropagation

After calculating the loss, we need to determine:

> Which weights caused the error and how should they change?

This is done using **backpropagation**.

Backpropagation moves information from:

```text
Output
  ↓
Hidden Layers
  ↓
Input
```

It calculates gradients.

```text
Forward Propagation:

Input → Hidden → Output → Loss

Backpropagation:

Loss → Output → Hidden → Input
```

---

# 18. What is a Gradient?

A gradient tells us:

> How much the loss changes when a parameter changes.

For a weight:

```text
∂Loss / ∂Weight
```

This tells us the direction in which the weight should change to reduce the loss.

---

# 19. Weight Update

Once we have the gradient, we update the weights.

Basic Gradient Descent:

```text
new_weight = old_weight - learning_rate × gradient
```

Mathematically:

```text
w = w - η ∂L/∂w
```

where:

```text
w = weight
η = learning rate
L = loss
```

The goal is:

```text
Reduce Loss
     ↓
Improve Predictions
```

---

# 20. Complete Training Cycle

An ANN learns using this cycle:

```text
       Input
         ↓
 Forward Propagation
         ↓
     Prediction
         ↓
    Loss Function
         ↓
   Calculate Loss
         ↓
   Backpropagation
         ↓
     Gradients
         ↓
    Weight Update
         ↓
       Repeat
```

This happens many times during training.

---

# 21. Activation Functions

Activation functions introduce **non-linearity** into neural networks.

Without activation functions, stacking many layers would essentially behave like one linear transformation.

Activation functions allow neural networks to learn complex patterns.

Common activation functions:

```text
Sigmoid
Tanh
ReLU
Leaky ReLU
Parametric ReLU
ELU
Softmax
```

---

# 22. Sigmoid

Formula:

```text
σ(z) = 1 / (1 + e⁻ᶻ)
```

Output range:

```text
0 → 1
```

Shape:

```text
       1 |          ______
         |       __/
         |     _/
       0.5|----/
         |  __/
         |/
       0 |________________
             z
```

Useful for:

```text
Binary classification
```

Example:

```text
0.87 → 87% probability
```

### Problem

Sigmoid can suffer from the **vanishing gradient problem** when values become very large or very small.

---

# 23. Tanh

Formula:

```text
tanh(z)
```

Output range:

```text
-1 → +1
```

Shape:

```text
      +1 |       ______
         |     /
       0 |----/
         |  /
      -1 |__/
         |____________
```

Compared with sigmoid:

```text
Sigmoid → 0 to 1

Tanh → -1 to +1
```

Tanh is zero-centered.

### Problem

It can also suffer from vanishing gradients.

---

# 24. ReLU

ReLU stands for:

**Rectified Linear Unit**

Formula:

```text
ReLU(z) = max(0,z)
```

Meaning:

```text
if z < 0 → 0

if z > 0 → z
```

Example:

```text
z = -5 → 0

z = 3 → 3
```

Visual:

```text
       y
       |
       |        /
       |       /
       |      /
_______|_____/
       |
       |
       +------------ x
```

### Advantages

- Simple
- Fast
- Computationally efficient
- Helps reduce vanishing-gradient problems for positive inputs
- Very common in hidden layers

### Problem

**Dying ReLU**

If a neuron keeps producing negative values, its output becomes:

```text
0
```

and its gradient can become zero.

---

# 25. Leaky ReLU

Leaky ReLU tries to fix the dying ReLU problem.

Formula:

```text
f(x) =
{
x       if x > 0
αx      if x ≤ 0
}
```

where `α` is a small value.

Example:

```text
α = 0.01
```

Then:

```text
x = -5

Leaky ReLU = -0.05
```

Instead of:

```text
ReLU(-5) = 0
```

So negative values still have a small gradient.

---

# 26. Parametric ReLU (PReLU)

PReLU is similar to Leaky ReLU.

The difference is:

```text
Leaky ReLU → α is usually fixed

PReLU → α is learned during training
```

Formula:

```text
f(x) =
{
x       if x > 0
αx      if x ≤ 0
}
```

But in PReLU:

```text
α = learnable parameter
```

---

# 27. ELU

ELU stands for:

**Exponential Linear Unit**

For positive values:

```text
ELU(x) = x
```

For negative values:

```text
ELU(x) = α(eˣ - 1)
```

ELU provides a smooth curve for negative values.

Advantages:

- Helps avoid dead neurons
- Allows negative outputs
- Can produce smoother optimization

---

# 28. Softmax

Softmax is mainly used in the **output layer of multiclass classification**.

Suppose we have:

```text
Naruto
Sasuke
Sakura
```

The network produces raw scores:

```text
Naruto = 2.5
Sasuke = 1.5
Sakura = 0.5
```

Softmax converts them into probabilities:

```text
Naruto = 0.67
Sasuke = 0.24
Sakura = 0.09
```

The probabilities add up to:

```text
1
```

or:

```text
100%
```

Formula:

```text
softmax(zᵢ) = eᶻⁱ / Σeᶻʲ
```

---

# 29. Activation Function Summary

| Activation | Range | Common Use |
|---|---:|---|
| Sigmoid | 0 to 1 | Binary output |
| Tanh | -1 to 1 | Older hidden layers / some sequence models |
| ReLU | 0 to ∞ | Hidden layers |
| Leaky ReLU | -∞ to ∞ | Hidden layers |
| PReLU | -∞ to ∞ | Hidden layers |
| ELU | -α to ∞ | Hidden layers |
| Softmax | 0 to 1 | Multiclass output |

---

# 30. Vanishing Gradient Problem

During backpropagation, gradients are passed backward through many layers.

Sometimes gradients become extremely small.

```text
Large gradient
     ↓
Smaller
     ↓
Smaller
     ↓
Very small
     ↓
Almost zero
```

Then earlier layers learn extremely slowly.

This is called:

**Vanishing Gradient Problem**

It was especially associated with saturating activations such as:

```text
Sigmoid
Tanh
```

---

# 31. Why Does Vanishing Gradient Happen?

During backpropagation, gradients are multiplied through layers.

If many values are smaller than 1:

```text
0.5 × 0.5 × 0.5 × 0.5 × ...
```

the result becomes extremely small.

Therefore:

```text
Gradient → 0
```

and:

```text
Weight updates → very small
```

So learning becomes slow.

---

# 32. Exploding Gradient Problem

The opposite problem is:

**Exploding Gradient**

Gradients become extremely large.

```text
Small
  ↓
Large
  ↓
Very large
  ↓
Huge
```

Then weight updates can become enormous.

This can make training unstable.

---

# 33. Gradient Problems

```text
Gradient Problems
       │
       ├── Vanishing Gradient
       │       ↓
       │   Gradient becomes too small
       │
       └── Exploding Gradient
               ↓
          Gradient becomes too large
```

Common techniques to reduce these problems include:

```text
Better activation functions
Better weight initialization
Batch normalization
Appropriate learning rate
Gradient clipping
```

---

# 34. Optimizers

An **optimizer** decides how the model's weights should be updated using gradients.

Basic idea:

```text
Loss
 ↓
Gradient
 ↓
Optimizer
 ↓
New Weights
```

Common optimizers:

```text
Gradient Descent
SGD
Mini-Batch Gradient Descent
SGD + Momentum
AdaGrad
RMSProp
Adam
```

---

# 35. Gradient Descent

Basic Gradient Descent uses the entire training dataset to calculate the gradient.

```text
Entire Dataset
      ↓
Calculate Gradient
      ↓
Update Weights
```

Formula:

```text
w = w - η∇L
```

### Advantage

Stable gradient estimate.

### Disadvantage

Can be slow for large datasets.

---

# 36. Stochastic Gradient Descent (SGD)

SGD updates weights using **one training example at a time**.

```text
Dataset
   ↓
Example 1 → Update
Example 2 → Update
Example 3 → Update
Example 4 → Update
```

Instead of waiting for the entire dataset.

### Advantage

- Faster updates
- Can escape some local regions because of noisy updates

### Disadvantage

- Noisy training
- Loss can fluctuate

---

# 37. Mini-Batch Gradient Descent

Mini-batch gradient descent uses a small group of examples.

For example:

```text
Dataset = 10,000 examples

Batch size = 32
```

Training:

```text
Examples 1-32   → Update
Examples 33-64  → Update
Examples 65-96  → Update
...
```

This is extremely common in neural-network training.

---

# 38. Batch Gradient Descent vs SGD vs Mini-Batch

| Method | Data used per update |
|---|---|
| Batch GD | Entire dataset |
| SGD | 1 example |
| Mini-Batch GD | Small batch |

Visual:

```text
Batch GD:

[████████████████████] → Update


SGD:

[█] → Update
[█] → Update
[█] → Update


Mini-Batch:

[████] → Update
[████] → Update
[████] → Update
```

---

# 39. SGD with Momentum

Momentum helps the optimizer move in a more consistent direction.

Instead of considering only the current gradient:

```text
Current Gradient
```

it also remembers previous updates.

Conceptually:

```text
Previous direction
       +
Current gradient
       ↓
Momentum
       ↓
Weight update
```

This can:

- Speed up convergence
- Reduce oscillations
- Help move through shallow regions

---

# 40. AdaGrad

AdaGrad gives different learning rates to different parameters.

Frequently updated parameters:

```text
smaller learning rate
```

Rarely updated parameters:

```text
larger learning rate
```

This can be useful for sparse data.

### Problem

The learning rate can keep decreasing and eventually become extremely small.

---

# 41. RMSProp

RMSProp was designed to address the aggressive learning-rate decay problem of AdaGrad.

It maintains an exponentially weighted average of squared gradients.

Conceptually:

```text
Recent gradients
       ↓
Squared gradients
       ↓
Moving average
       ↓
Adaptive learning rate
```

It generally performs better than AdaGrad on many neural-network problems.

---

# 42. Adam

Adam stands for:

**Adaptive Moment Estimation**

Adam combines ideas from:

```text
Momentum
+
RMSProp
```

It keeps track of:

```text
1. First moment → average gradient
2. Second moment → average squared gradient
```

Then it uses these to adapt the learning rate.

Adam is one of the most commonly used optimizers.

Conceptually:

```text
Gradient
   │
   ├────────→ Momentum-like information
   │
   └────────→ RMSProp-like information
                  │
                  ↓
                 Adam
                  │
                  ↓
             Weight Update
```

---

# 43. Optimizer Summary

| Optimizer | Main Idea |
|---|---|
| Gradient Descent | Uses entire dataset |
| SGD | One example per update |
| Mini-Batch GD | Small batch per update |
| Momentum | Uses previous direction |
| AdaGrad | Adaptive learning rate |
| RMSProp | Moving average of squared gradients |
| Adam | Momentum + adaptive learning rate |

---

# 44. Weight Initialization

At the beginning, neural-network weights cannot all be the same.

Bad:

```text
w₁ = 0
w₂ = 0
w₃ = 0
w₄ = 0
```

If all neurons start identically, they can learn the same thing.

This is called a **symmetry problem**.

Therefore, we initialize weights carefully.

---

# 45. Common Weight Initialization Techniques

## Zero Initialization

```text
W = 0
```

Usually bad for hidden layers because neurons can remain symmetric.

---

## Random Initialization

Weights are randomly initialized.

Example:

```text
W = random values
```

This breaks symmetry.

However, purely random values may have poor scaling.

---

# 46. Xavier / Glorot Initialization

Xavier initialization chooses the scale of weights based on the number of input/output neurons.

It is commonly associated with:

```text
Sigmoid
Tanh
```

The goal is to keep activations and gradients at reasonable scales.

---

# 47. He Initialization

He initialization is commonly used with:

```text
ReLU
Leaky ReLU
```

It is designed to account for the behavior of ReLU-like activations.

A common form is:

```text
W ~ N(0, 2/n_in)
```

where:

```text
n_in = number of input neurons
```

---

# 48. Weight Initialization Summary

```text
Zero Initialization
        ↓
Usually bad for hidden layers

Random Initialization
        ↓
Breaks symmetry

Xavier Initialization
        ↓
Often good for Sigmoid/Tanh

He Initialization
        ↓
Often good for ReLU-family activations
```

---

# 49. Dropout

**Dropout** is a regularization technique used to reduce overfitting.

During training, dropout randomly turns off some neurons.

Example:

```text
Before Dropout:

○ ─── ○ ─── ○ ─── ○
│     │     │     │
○ ─── ○ ─── ○ ─── ○


After Dropout:

○ ─── X ─── ○ ─── ○
│           │
X ─── ○ ─── X ─── ○
```

`X` means the neuron is temporarily ignored for that training step.

---

# 50. Why Do We Use Dropout?

Without dropout, the network can become too dependent on particular neurons.

It may memorize the training data.

This causes:

```text
Training accuracy → Very high
Validation accuracy → Lower
```

which is:

**Overfitting**

Dropout forces the network to learn more robust features.

---

# 51. Important Dropout Point

Dropout is mainly applied during:

```text
Training
```

During inference/testing:

```text
Dropout is turned OFF
```

The complete network is used.

---

# 52. Complete ANN Learning Process

Now let's combine everything.

```text
                  INPUT
                    │
                    ↓
             ┌─────────────┐
             │ Hidden Layer│
             │             │
             │ Wx + b      │
             │     ↓       │
             │ Activation  │
             └──────┬──────┘
                    │
                    ↓
             ┌─────────────┐
             │ Hidden Layer│
             │             │
             │ Wx + b      │
             │     ↓       │
             │ Activation  │
             └──────┬──────┘
                    │
                    ↓
                  OUTPUT
                    │
                    ↓
                Prediction
                    │
                    ↓
              Loss Function
                    │
                    ↓
             Backpropagation
                    │
                    ↓
                 Gradients
                    │
                    ↓
                Optimizer
                    │
                    ↓
              Weight Update
                    │
                    └──────────→ Repeat
```

# 53. One Complete Example — Naruto Classification 🍥

Let's imagine we build an ANN that predicts whether a Naruto character belongs to a particular class based on three numerical features.

Suppose our features are:

```text
x₁ = Chakra level
x₂ = Speed
x₃ = Battle experience
```

Our target:

```text
1 = Strong Shinobi
0 = Not Strong Shinobi
```

---

# 54. Step 1 — Input

Suppose Naruto has:

```text
Chakra = 9
Speed = 8
Experience = 7
```

Therefore:

```text
X = [9, 8, 7]
```

The input layer receives:

```text
       Input Layer

          x₁ = 9
             \
          x₂ = 8 ─────→ Hidden Layer
             /
          x₃ = 7
```

---

# 55. Step 2 — Hidden Layer

Suppose one hidden neuron has:

```text
w₁ = 0.2
w₂ = 0.4
w₃ = 0.3

b = 0.5
```

The neuron calculates:

```text
z = w₁x₁ + w₂x₂ + w₃x₃ + b
```

Substitute:

```text
z = (0.2)(9) + (0.4)(8) + (0.3)(7) + 0.5
```

```text
z = 1.8 + 3.2 + 2.1 + 0.5
```

```text
z = 7.6
```

Then:

```text
a = ReLU(7.6)
```

Therefore:

```text
a = 7.6
```

---

# 56. Step 3 — More Hidden Neurons

A real hidden layer contains many neurons:

```text
                 Hidden Layer

x₁ ─────────────→ ○
x₂ ─────────────→ ○
x₃ ─────────────→ ○
                  │
                  ├──→ ○
                  │
                  └──→ ○
```

Each neuron has its own:

```text
weights
+
bias
+
activation function
```

So they can learn different patterns.

---

# 57. Step 4 — Output Layer

For binary classification, we can use:

```text
Sigmoid
```

The output might be:

```text
ŷ = 0.94
```

Meaning:

```text
94% probability → Strong Shinobi
```

Therefore:

```text
Prediction = 1
```

---

# 58. Complete Forward Pass

```text
Naruto Features

[9, 8, 7]
    │
    ↓
┌─────────────────┐
│ Hidden Layer    │
│                 │
│ z = Wx + b      │
│       ↓         │
│      ReLU       │
└────────┬────────┘
         │
         ↓
┌─────────────────┐
│ Output Layer    │
│                 │
│ z = Wx + b      │
│       ↓         │
│    Sigmoid      │
└────────┬────────┘
         │
         ↓
       0.94
         │
         ↓
   Strong Shinobi
```

---

# 59. Step 5 — Calculate Loss

Suppose the actual label is:

```text
y = 1
```

and the model predicts:

```text
ŷ = 0.94
```

The prediction is close to the correct answer.

Therefore:

```text
Loss = small
```

If instead:

```text
ŷ = 0.10
```

then:

```text
Loss = large
```

---

# 60. Step 6 — Backpropagation

Now the network asks:

> Which weights contributed to this error?

Backpropagation calculates gradients.

```text
Loss
 ↓
Output Layer
 ↓
Hidden Layer
 ↓
Weights
```

For example:

```text
∂Loss/∂w₁
∂Loss/∂w₂
∂Loss/∂w₃
```

These gradients tell us how each weight should change.

---

# 61. Step 7 — Optimizer Updates Weights

Suppose:

```text
old weight = 0.20

gradient = 0.10

learning rate = 0.01
```

Gradient descent:

```text
new weight = old weight - learning_rate × gradient
```

Therefore:

```text
new weight = 0.20 - (0.01)(0.10)

new weight = 0.199
```

The weight has been slightly adjusted.

---

# 62. Step 8 — Repeat

The model performs this process again and again:

```text
Input
 ↓
Forward Propagation
 ↓
Prediction
 ↓
Loss
 ↓
Backpropagation
 ↓
Gradients
 ↓
Optimizer
 ↓
Weight Update
 ↓
Forward Propagation again
 ↓
...
```

Eventually the network learns weights that produce better predictions.

---

# 63. Where Does Dropout Fit?

During training:

```text
Input
 ↓
Hidden Layer
 ↓
Dropout
 ↓
Next Hidden Layer
 ↓
Output
```

Example:

```text
Hidden neurons:

○ ○ ○ ○ ○ ○ ○ ○

Dropout:

○ X ○ ○ X ○ X ○

X = temporarily disabled
```

The disabled neurons are randomly selected during training.

---

# 64. Where Does Weight Initialization Fit?

Weight initialization happens **before training**.

# 66. Quick Revision

## Perceptron

```text
z = Wx + b
```

then:

```text
a = activation(z)
```

---

## ANN

```text
Input Layer
     ↓
Hidden Layer
     ↓
Hidden Layer
     ↓
Output Layer
```

---

## Forward Propagation

```text
Input → Output
```

Used to generate prediction.

---

## Loss

```text
Prediction vs Actual
```

Measures how wrong the model is.

---

## Backpropagation

```text
Loss → Backward
```

Calculates gradients.

---

## Weight Update

```text
w = w - η∇L
```

Updates the weights to reduce loss.

---

## Activation Functions

```text
Sigmoid
Tanh
ReLU
Leaky ReLU
PReLU
ELU
Softmax
```

---

## Vanishing Gradient

```text
Gradient → Very Small
```

Learning becomes extremely slow.

---

## Exploding Gradient

```text
Gradient → Very Large
```

Training becomes unstable.

---

## Optimizers

```text
Gradient Descent
SGD
Mini-Batch GD
Momentum
AdaGrad
RMSProp
Adam
```

---

## Weight Initialization

```text
Random
Xavier
He
```

The goal is to start weights at useful scales and avoid symmetry / unstable activations and gradients.

---

## Dropout

```text
Randomly deactivate neurons during training
```

Main purpose:

```text
Reduce Overfitting
```

And the purpose of everything is ultimately:

```text
                    Reduce Loss
                        ↓
                 Better Weights
                        ↓
                Better Predictions
```
