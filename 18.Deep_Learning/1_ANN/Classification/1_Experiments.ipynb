````python
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
import dill
````

___
# Load The Dataset 
___

````python
data = pd.read_csv("Churn_Modelling.csv")
data
````

```
    Predict Whether Customer will Exist/Leave the Bank or not 

___

# PreProcessing Data
___

## 1 Drop Irrelavent columns 

```
RowNumber, CutomerId, Surname

````python
data = data.drop(["RowNumber", "CustomerId", "Surname"], axis = 1)
````

````python
data
````

```
Geography & Gneder are Categorical Featutes Encode them

````python
label_encoder_gender = LabelEncoder()
data["Gender"] = label_encoder_gender.fit_transform(data["Gender"])
````

````python
data
````

````python
# use OHE for Geography Feature

from sklearn.preprocessing import OneHotEncoder

ohe_encoder_geo = OneHotEncoder(sparse_output=False, handle_unknown="ignore")

geo_encoded = ohe_encoder_geo.fit_transform(data[["Geography"]])

````

````python
ohe_encoder_geo.get_feature_names_out()
````

````python
geo_encoded_df = pd.DataFrame(
    geo_encoded,
    columns=ohe_encoder_geo.get_feature_names_out()
)
````

````python
# print Geo encoded Df 
geo_encoded_df
````

### Combne OHE cols with Og Data

````python
data = pd.concat(
    [data, geo_encoded_df],
    axis=1
)
````

````python
data
````

`` Drop the Geography col

````python
data = data.drop("Geography", axis=1)
````

````python
data
````
