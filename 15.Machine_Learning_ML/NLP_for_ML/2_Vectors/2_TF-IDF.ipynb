# TF-IDF (Term Frequency – Inverse Document Frequency)

## 1. What is TF-IDF?

**TF-IDF** is a technique used in NLP to convert text into **numerical values**.

It gives a word a score based on:

1. **How frequently the word appears in a document**
2. **How rare or common the word is across all documents**

The main intuition is:

> **A word is important if it appears frequently in one document but does not appear in many other documents.**

---

# 2. Why Do We Need TF-IDF?

Suppose we have these documents:

```text
Document 1:
I love machine learning

Document 2:
I love deep learning

Document 3:
I love Python programming
```

The word:

```text
"love"
```

appears in almost every document.

So `"love"` doesn't help us distinguish between the documents very much.

But:

```text
"machine"
"deep"
"Python"
"programming"
```

are more specific to particular documents.

TF-IDF tries to give:

```text
Common words      → Lower importance
Rare/specific words → Higher importance
```

---

# 3. The Intuition

Think about reading many documents.

Suppose the word:

```text
"the"
```

appears everywhere.

It doesn't tell you much about what the document is about.

So TF-IDF gives it a **low score**.

Now suppose:

```text
"neuralnetwork"
```

appears many times in one document but almost nowhere else.

That word is probably important for that document.

So TF-IDF gives it a **higher score**.

---

# 4. TF-IDF Has Two Parts

TF-IDF:

```text
TF-IDF = TF × IDF
```

Where:

```text
TF  = Term Frequency
IDF = Inverse Document Frequency
```

---

# 5. Term Frequency (TF)

TF tells us:

> **How frequently does a word occur inside a document?**

A basic formula is:

```text
TF = Number of times term appears in document
     -----------------------------------------
          Total number of words in document
```

For example:

```text
Document:

"I love Python Python Python"
```

Total words:

```text
5
```

`Python` occurs:

```text
3 times
```

Therefore:

```text
TF(Python) = 3 / 5
           = 0.6
```

So the word has a relatively high TF.

---

# 6. Inverse Document Frequency (IDF)

IDF asks:

> **How rare is this word across all documents?**

If a word appears in almost every document, it isn't very useful for distinguishing documents.

Therefore:

```text
Common word → Low IDF
Rare word   → High IDF
```

A common formula is:

```text
IDF = log(N / df)
```

Where:

```text
N  = Total number of documents
df = Number of documents containing the word
```

---

# 7. Simple IDF Example

Suppose we have:

```text
N = 10 documents
```

The word:

```text
"Python"
```

appears in:

```text
2 documents
```

Therefore:

```text
df = 2
```

IDF:

```text
IDF = log(10 / 2)

    = log(5)
```

This gives a relatively high value.

---

Now suppose:

```text
"the"
```

appears in:

```text
10 documents
```

Then:

```text
df = 10
```

Therefore:

```text
IDF = log(10 / 10)

    = log(1)

    = 0
```

So `"the"` gets an IDF of approximately:

```text
0
```

This makes sense because `"the"` doesn't help distinguish documents.

---

# 8. TF-IDF

Finally:

```text
TF-IDF = TF × IDF
```

Suppose:

```text
TF = 0.6

IDF = 1.5
```

Then:

```text
TF-IDF = 0.6 × 1.5

       = 0.9
```

The word receives a relatively high importance score.

---

# 9. The Main Logic

Remember this:

```text
                TF-IDF
                   │
          ┌────────┴────────┐
          ↓                 ↓
         TF                IDF
          │                 │
          ↓                 ↓
How often in       How rare across
this document?       all documents?
          │                 │
          └────────┬────────┘
                   ↓
              TF × IDF
                   ↓
             Word Score
```

---

# 10. Very Important Intuition

TF-IDF rewards a word when:

```text
The word appears frequently
in the current document
          +
The word is rare across
the entire collection
```

So:

```text
High TF + High IDF
        ↓
High TF-IDF
        ↓
Important word
```

While:

```text
High TF + Low IDF
        ↓
Low TF-IDF
        ↓
Common word
```

---

Suppose we have:

```text
Document 1:
Python Python Python is great

Document 2:
Java is great

Document 3:
Java is great
```

Consider the word:

```text
Python
```

### TF

In Document 1:

```text

```

### IDF

Python only appears in:

```text

```

Therefore:

```text

   ↓

```

Now consider:

```text

```

It appears in multiple documents.

Therefore:

```text
High/medium TF
      +
Low IDF
      ↓
Low TF-IDF
```

# 12. TF-IDF vs Bag of Words

## Bag of Words

BoW mainly asks:

> How many times does this word appear?

Example:

```text
Python → 3
Java   → 0
Machine → 1
```

It doesn't consider how common the word is across the whole dataset.

---

## TF-IDF

TF-IDF asks:

> How frequently does this word appear here, and how rare is it across all documents?

So:

```text
BoW:
word frequency

TF-IDF:
word frequency × word importance
```

---

# 13. Example Comparison

Suppose:

```text
Document 1:
Python is amazing

Document 2:
Python is easy

Document 3:
Python is powerful
```

The word:

```text
Python
```

appears in every document.

So although it appears in each document, it doesn't help distinguish them.

TF-IDF reduces its importance.

But:

```text
amazing
easy
powerful
```

appear in only one document each.

They receive higher importance in their respective documents.

---

# 14. TF-IDF Creates Numerical Vectors

Suppose our vocabulary is:

```text
["python", "machine", "learning", "java"]
```

A document might become:

```text
[0.8, 0.0, 0.5, 0.0]
```

Another document:

```text
[0.0, 0.6, 0.7, 0.4]
```

So text becomes numerical data.

This allows us to use traditional Machine Learning models.

---

# 16. Complete Basic Code

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "I love machine learning",
    "I love deep learning",
    "Python is powerful"
]

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(documents)

print("Vocabulary:")
print(vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(X.toarray())
```

---

# 17. `fit_transform()`

This:

```python
X = vectorizer.fit_transform(documents)
```

does two things.

### `fit()`

Learns the vocabulary and calculates the required statistics from the training documents.

### `transform()`

Converts the documents into TF-IDF numerical vectors.

Therefore:

```python
fit_transform()
```

means:

```text
Learn from text
      +
Convert text → TF-IDF vectors
```

---

For example:


---

## 3. Highlights Important Words

Words that are:


# 20. Disadvantages of TF-IDF

## 1. Does Not Understand Word Meaning

TF-IDF doesn't understand that:

```text
car
automobile
vehicle
```

have related meanings.

It treats them as different features.

---

## 2. Ignores Word Order

Consider:

```text
"I love Python"
```

and:

```text
"Python love I"
```

Without n-grams, TF-IDF doesn't understand the order of the words.

---

## 3. No Deep Context Understanding

Consider:

```text
"I went to the bank to deposit money."

"I sat beside the bank of the river."
```

The word:

```text
bank
```

has different meanings.

TF-IDF doesn't understand that contextual difference.

---

## 4. Large Feature Space

If your dataset contains thousands or millions of unique words, TF-IDF can create a huge number of features.

For example:

```text
1,000,000 unique words
        ↓
Potentially
1,000,000 features
```

---

## 5. Sparse Representation

Most documents contain only a small percentage of the entire vocabulary.

Therefore the resulting matrix can contain many zeros:

```text
[0, 0, 0, 0.7, 0, 0, 0.2, 0, ...]
```

This is called a **sparse matrix**.

---

## 6. Doesn't Understand Synonyms

For example:

```text
good
great
excellent
```

are treated as separate words/features.

TF-IDF doesn't inherently know that their meanings are related.

---

# 21. Bag of Words vs TF-IDF

| Feature | Bag of Words | TF-IDF |
|---|---|---|
| Converts text to numbers | ✅ | ✅ |
| Counts word frequency | ✅ | ✅ |
| Considers word rarity | ❌ | ✅ |
| Reduces common-word importance | ❌ | ✅ |
| Understands word meaning | ❌ | ❌ |
| Understands context | ❌ | ❌ |
| Understands word order | ❌ | ❌ |
| Can use n-grams | ✅ | ✅ |
| Works with traditional ML | ✅ | ✅ |

# 22. Mental Model

Think of TF-IDF as asking two questions:

```text
Question 1:

"How often does this word appear
in THIS document?"

             ↓

             TF
```

and:

```text
Question 2:

"How common is this word

             ↓

             IDF
```

Then:

```text
TF × IDF
   ↓
TF-IDF score
```

---

# 23. The Most Important Intuition

Remember:

```text
Word appears frequently
in this document
        +
Word is rare across documents
        ↓
HIGH TF-IDF
```

But:

```text
Word appears frequently
in this document
        +
Word appears everywhere
        ↓
LOW TF-IDF
```

So the main idea is:

> **TF-IDF doesn't simply ask "How often does this word occur?" It asks "How useful is this word for distinguishing this document from the others?"**

````python
import pandas as pd 

messages = pd.read_csv("SMSSpamCollection", sep="\t", names = ["label", "Message"])
messages
````

# cleaning and Text Proccessing 
___

````python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re

lemmitizer = WordNetLemmatizer()
````

````python
corpus = [ ]

for i in range(0, len(messages)):

    review = re.sub('[^a-z^]', " ", messages["Message"][i])
    review = review.lower()

    review = review.split()

    review = [lemmitizer.lemmatize(word) for word in review if word not in stopwords.words("english")]

    review = " ".join(review)

    corpus.append(review)
````

````python
vectorizer = TfidfVectorizer(max_features=100)

X = vectorizer.fit_transform(corpus).toarray()

X
````
