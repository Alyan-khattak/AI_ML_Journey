# Tokenization using NLTK

Tokenization is the process of breaking text into smaller pieces called **tokens**.

For example:

```text
"I love Machine Learning!"
```

Can become:

```text
["I", "love", "Machine", "Learning", "!"]
```

# 2. Corpus

A **corpus** is a collection of text/documents used for NLP.

Example:

```python
corpus = """
Machine Learning is amazing.
Natural Language Processing works with text.
I am learning NLP.
"""
```

Think:

```text
Corpus
│
├── Sentence 1
├── Sentence 2
└── Sentence 3
```

A corpus can contain:

- One document
- Many documents
- Thousands or millions of documents

---

# 3. Sentence Tokenization

Sentence tokenization means:

> Breaking a paragraph into individual sentences.

Example:

```python
from nltk.tokenize import sent_tokenize

text = """
Machine Learning is amazing.
I am learning NLP.
NLTK is useful for text processing.
"""

sentences = sent_tokenize(text)

print(sentences)
```

Output:

```python
[
    "Machine Learning is amazing.",
    "I am learning NLP.",
    "NLTK is useful for text processing."
]
```

So:

```text
Paragraph
    ↓
Sentence Tokenization
    ↓
Sentence 1
Sentence 2
Sentence 3
```

---

# 4. Word Tokenization

Word tokenization means:

> Breaking a sentence into individual words/tokens.

Example:

```python
from nltk.tokenize import word_tokenize

text = "I am learning Natural Language Processing."

words = word_tokenize(text)

print(words)
```

Output:

```python
[
    "I",
    "am",
    "learning",
    "Natural",
    "Language",
    "Processing",
    "."
]
```

Notice that punctuation can also become a token:

```text
"." → token
```

---

# 5. Sentence vs Word Tokenization

### Sentence Tokenization

```text
"I love NLP. NLP is interesting."
                ↓
        Sentence Tokenization
                ↓
[
    "I love NLP.",
    "NLP is interesting."
]
```

### Word Tokenization

```text
"I love NLP."
      ↓
Word Tokenization
      ↓
["I", "love", "NLP", "."]
```

# 6. Vocabulary

A **vocabulary** is the collection of unique words/tokens found in the text.

Example:

```python
text = "I love NLP and I love Python."
```

Word tokens:

```python
["I", "love", "NLP", "and", "I", "love", "Python", "."]
```

Unique tokens:

```python
["I", "love", "NLP", "and", "Python", "."]
```

We can create a vocabulary using:

```python
words = word_tokenize(text)

vocabulary = set(words)

print(vocabulary)
```

---

# 7. Vocabulary Size

To find the number of unique tokens:

```python
vocab_size = len(set(words))

print(vocab_size)
```

Example:

```text
Total tokens   → 8
Unique tokens  → 6

Vocabulary size → 6
```

---

# 8. Complete Example

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

text = """
Machine Learning is amazing.
I am learning Natural Language Processing.
NLP works with text.
"""

# Sentence tokenization
sentences = sent_tokenize(text)

# Word tokenization
words = word_tokenize(text)

# Vocabulary
vocabulary = set(words)

print("Sentences:")
print(sentences)

print("\nWords:")
print(words)

print("\nVocabulary:")
print(vocabulary)

print("\nVocabulary Size:")
print(len(vocabulary))
```

# 9. Important Terms

| Term | Meaning |
|---|---|
| **Corpus** | Collection of text/documents |
| **Sentence** | A complete textual statement |

| **Word Tokenization** | Splitting text into words/tokens |
| **Sentence Tokenization** | Splitting text into sentences |
| **Vocabulary** | Unique tokens in the text |
| **Vocabulary Size** | Number of unique tokens |

# Treebank Tokenizer

`TreebankWordTokenizer` is an NLTK tokenizer that splits text into **words and punctuation tokens** using rules based on the Penn Treebank tokenization conventions.

It is useful for **word-level tokenization**.

---

## 1. Import

```python
from nltk.tokenize import TreebankWordTokenizer
```

---

## 2. Basic Syntax

```python
tokenizer = TreebankWordTokenizer()

tokens = tokenizer.tokenize(text)

print(tokens)
```

---

## 3. Example

```python
from nltk.tokenize import TreebankWordTokenizer

text = "I am learning NLP. It's amazing!"

tokenizer = TreebankWordTokenizer()

tokens = tokenizer.tokenize(text)

print(tokens)
```

Output:

```python
[
    "I",
    "am",
    "learning",
    "NLP.",
    "It",
    "'s",
    "amazing",
    "!"
]
```

---

## 4. Why Use It?

It handles things like:

- Words
- Punctuation
- Contractions
- Quotes
- Special characters

Example:

```text
"Don't do this!"
```

Can be tokenized into something like:

```python
["Do", "n't", "do", "this", "!"]
```

---

## 5. Short Version

```python
from nltk.tokenize import TreebankWordTokenizer

tokens = TreebankWordTokenizer().tokenize(corpus)

print(tokens)
```

### Remember

  ↓
