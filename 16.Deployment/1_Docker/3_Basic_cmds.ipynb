# 1. Check Docker Version

## What is it for?

Checks whether Docker is installed and shows the Docker version.

## Syntax

```bash
docker --version
```

or:

```bash
docker version
```

### Difference

```bash
docker --version
```

Shows a short version:

```text
Docker version 28.x.x
```

While:

```bash
docker version
```

shows more detailed information about the Docker Client and Server.

# 4. List Docker Images

## What is it for?

Shows all Docker images stored locally on your machine.

## Syntax

```bash
docker images
```

Example:

```bash
docker images
```

Output may look like:

```text
IMAGE                           ID             DISK USAGE
docker/getting-started:latest   d79336f4812b   73.9MB
hello-world:latest              96498ffd522e   25.9kB
n8nio/n8n:latest                cf11c96b0d00    2.47GB
```

```text
IMAGE
```

Name of the image.

```text
ID
```

Unique ID of the image.

```text
DISK USAGE
```

How much disk space the image uses.

# 5. Pull an Image

## What is it for?

Downloads an image from a Docker registry, usually **Docker Hub**, to your computer.

## Syntax

```bash
docker pull IMAGE_NAME
```

Example:

```bash
docker pull nginx
```

This downloads the `nginx` image.

You can also specify a version/tag:

```bash
docker pull nginx:latest
```

```bash
docker pull python:3.12
```

```bash
docker pull ubuntu:24.04
```

### Breakdown

```bash
docker pull nginx:latest
│      │    │     │
│      │    │     └── Tag / version
│      │    └──────── Image name
│      └───────────── Pull command
└──────────────────── Docker
```

# 6. Remove an Image

## What is it for?

Deletes a Docker image from your local machine.

## Syntax

```bash
docker rmi IMAGE_NAME
```

or:

```bash
docker rmi IMAGE_ID
```

Example:

```bash
docker rmi hello-world
```

or:

```bash
docker rmi 96498ffd522e
```

### Important

`rmi` means:

```text
remove image
```

It does **not** mean remove container.

---

# 7. Force Remove an Image

## What is it for?

Forces Docker to remove an image.

## Syntax

```bash
docker rmi -f IMAGE
```

Example:

```bash
docker rmi -f hello-world
```

### `-f`

Means:

```text
force
```

Use it carefully.

# 📦 Docker Containers

A **container** is a running or stopped instance created from a Docker image.

```text
Docker Image
     │
     │ docker run
     ↓
Docker Container
```

---

# 8. Create and Run a Container

## What is it for?

Creates a container from an image and starts it.

## Syntax

```bash
docker run IMAGE
```

Example:

```bash
docker run hello-world
```

Docker will:

```text
Find image locally
      ↓
If not available
      ↓
Pull image
      ↓
Create container
      ↓
Start container
```

---

# 9. Run Container in Detached Mode

## What is it for?

Runs the container in the background.

## Syntax

```bash
docker run -d IMAGE
```

Example:

```bash
docker run -d nginx
```

### `-d`

Means:

```text
detached mode
```

The container runs in the background and your terminal is immediately available again.

---

# 10. Run Container with Port Mapping

## What is it for?

Connects a port on your computer to a port inside the container.

## Syntax

```bash
docker run -p HOST_PORT:CONTAINER_PORT IMAGE
```

Example:

```bash
docker run -p 80:80 nginx
```

### Breakdown

```text
-p 80:80
   │  │
   │  └── Container port
   │
   └───── Host port
```

So:

```text
Your Computer
localhost:80
     │
     ↓
Docker
     │
     ↓
Container:80
```

---

# 11. Run in Detached Mode + Port Mapping

A very common command:

```bash
docker run -d -p 80:80 nginx
```

### Breakdown

```text
docker
```

Docker command.

```text
run
```

Create and start a container.

```text
-d
```

Run in detached/background mode.

```text
-p 80:80
```

Map:

```text
Host port 80
      ↓
Container port 80
```

```text
nginx
```

Image to use.

So the complete meaning is:

> Create an Nginx container, run it in the background, and make container port `80` accessible through host port `80`.

# ⚠️ Important: Container ID vs Image ID

This is a very common beginner mistake.

Suppose:

```text
docker images
```

shows:

```text
IMAGE            ID
hello-world      96498ffd522e
```

That:

```text
96498ffd522e
```

is an **IMAGE ID**.

It cannot be used with:

```bash
docker rm
```

because `docker rm` removes containers.

You tried:

```bash
docker rm -f 96498ffd522e
```

and received:

```text
No such container
```

because:

```text
96498ffd522e
       ↓
Image ID
```

not:

```text
Container ID
```

To remove that image:

```bash
docker rmi 96498ffd522e
```

---

# 21. Open a Shell Inside a Container

A very useful command.

## Syntax

```bash
docker exec -it CONTAINER_ID bash
```

Example:

```bash
docker exec -it mycontainer bash
```

If the container doesn't have `bash`, try:

```bash
docker exec -it mycontainer sh
```

### Breakdown

```text
-it
│ │
│ └── Interactive
└──── Allocate a terminal
```

This allows you to interact with the container's shell.

Example:

```text
Host Terminal
     │
     │ docker exec -it ...
     ↓
Container Shell
     │
     ├── ls
     ├── cd
     ├── pwd
     └── etc.
```

# 🏗️ 25. Build an Image

## What is it for?

Builds a Docker image from a `Dockerfile`.

## Syntax

```bash
docker build -t IMAGE_NAME .
```

Example:

```bash
docker build -t myapp .
```

### Breakdown

```text
docker
```

Docker command.

```text
build
```

Build an image.

```text
-t myapp
```

Give the image a name/tag.

```text
.
```

Use the current directory as the build context.

So:

```bash
docker build -t myapp .
```

means:

> Build an image using the Dockerfile in the current directory and name the image `myapp`.

# 📤 27. Push an Image

## What is it for?

Uploads a Docker image to a Docker registry such as Docker Hub.

## Syntax

```bash
docker push IMAGE
```

Example:

```bash
docker push username/myapp:latest
```

Before pushing, you normally need to log in:

```bash
docker login
```

### Typical workflow

```text
Build Image
    ↓
Tag Image
    ↓
Login
    ↓
Push Image
    ↓
Docker Registry
```

Example:

```bash
docker build -t myapp .
```

```bash
docker tag myapp username/myapp:latest
```

```bash
docker login
```

```bash
docker push username/myapp:latest
```

---

# 📥 28. Pull an Image

Pull is the opposite of push.

```text
Push
Local Machine
     ↓
Docker Hub
```

```text
Pull
Docker Hub
     ↓
Local Machine
```

## Syntax

```bash
docker pull IMAGE
```

Example:

```bash
docker pull username/myapp:latest
```

# 🧹 31. Remove Unused Images

## What is it for?

Removes unused Docker images.

## Syntax

```bash
docker image prune
```

```bash
docker image prune -a
```

### Be careful

```bash
docker image prune -a
```

can remove images that you aren't currently using.

# 🧹 32. Docker System Cleanup

## What is it for?

Removes unused Docker resources.

## Syntax

```bash
docker system prune
```

More aggressive:

```bash
docker system prune -a
```

Be careful with this command because it can remove unused:

```text
Containers
Networks
Images
Build cache
```

# 🏷️ 33. Docker Image Tags

An image can have a tag.

```text
IMAGE:TAG
```

Example:

```bash
python:3.12
```

Here:

```text
python
  ↓
Image name

3.12
  ↓
Tag
```

Another example:

```bash
nginx:latest
```

```text
nginx
  ↓
Image

latest
  ↓
Tag
```

If you don't specify a tag:

```bash
docker pull nginx
```

Docker normally uses:

```text
nginx:latest
```

# 🔄 34. Complete Basic Docker Workflow

A very common workflow is:

```text
Dockerfile
     ↓
docker build
     ↓
Docker Image
     ↓
docker run
     ↓
Docker Container
     ↓
docker ps
     ↓
Check Container
     ↓
docker logs
     ↓
Check Logs
     ↓
docker stop
     ↓
Stop Container
     ↓
docker rm
     ↓
Remove Container
```

# 🚀 35. Simple Example

Suppose we have:

```text
myapp/
│
├── Dockerfile
├── requirements.txt
└── app.py
```

Build the image:

```bash
docker build -t myapp .
```

Check the image:

```bash
docker images
```

Run the container:

```bash
docker run -d -p 8000:8000 myapp
```

Check running containers:

```bash
docker ps
```

View logs:

```bash
docker logs CONTAINER_ID
```

Stop it:

```bash
docker stop CONTAINER_ID
```

Remove it:

```bash
docker rm CONTAINER_ID
```

Remove the image:

```bash
docker rmi myapp
```

# 🎯 Remember These 10 First

If you're just starting Docker, don't try to memorize everything at once.

Start with these:

```bash
docker --version
docker images
docker pull IMAGE
docker build -t NAME .
docker run IMAGE
docker ps
docker ps -a
docker stop CONTAINER
docker rm CONTAINER
docker rmi IMAGE
```

And remember the most important relationship:

```text
                 Dockerfile
                     ↓
              docker build
                     ↓
                🐳 Image
                     ↓
                docker run
                     ↓
              📦 Container
                     ↓
               Application
```

> **`docker rmi` → removes an IMAGE**

> **`docker rm` → removes a CONTAINER**

> **`docker pull` → downloads an IMAGE**

> **`docker push` → uploads an IMAGE**

> **`docker run` → creates and starts a CONTAINER from an IMAGE**

> **`docker start` → starts an already existing CONTAINER**
