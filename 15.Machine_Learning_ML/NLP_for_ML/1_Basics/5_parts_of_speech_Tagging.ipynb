# Parts of Speech (POS) Tagging

## 1. What is POS Tagging?

**Part of Speech (POS) tagging** means assigning a grammatical category to each word in a sentence.

Example:

```text
The cat is running.
```

POS tagging:

```text
The     → DT
cat     → NN
is      → VBZ
running → VBG
```

So the computer understands that:

```text
The     = Determiner
cat     = Noun
is      = Verb
running = Verb
```

---

# 2. Why Do We Use POS Tagging?

POS tagging helps an NLP system understand the **role of each word in a sentence**.

For example:

```text
I book a ticket.
```

Here:

```text
book → Verb
```

But:

```text
I read a book.
```

Here:

```text
book → Noun
```

The same word can have different meanings/roles depending on the sentence.

---

# 3. Common Parts of Speech

| POS | Meaning | Example |
|---|---|---|
| Noun | Person, place, thing | student |
| Verb | Action/state | run |
| Adjective | Describes noun | beautiful |
| Adverb | Describes verb/adjective | quickly |
| Pronoun | Replaces noun | he |
| Determiner | Specifies noun | the |
| Preposition | Relationship/location | in |
| Conjunction | Connects words | and |
| Interjection | Emotion | wow |

# 4. NLTK POS Tagging

NLTK provides:

```python
nltk.pos_tag()
```

Basic syntax:

```python
nltk.pos_tag(words)
```

# 5. Basic Example

```python
import nltk

nltk.download("averaged_perceptron_tagger_eng")

sentence = "The cat is running."

words = nltk.word_tokenize(sentence)

tags = nltk.pos_tag(words)

print(tags)
```

Possible output:

```text
[
    ('The', 'DT'),
    ('cat', 'NN'),
    ('is', 'VBZ'),
    ('running', 'VBG'),
    ('.', '.')
]
```

---

# 6. Understanding the Output

The output is in the form:

```text
(word, POS_tag)
```

For example:

```text
('cat', 'NN')
```

means:

```text
cat → Noun
```

And:

```text
('running', 'VBG')
```

means:

```text
running → Verb
```

# 7. Common NLTK POS Tags

### Nouns

```text
NN   → Noun, singular
NNS  → Noun, plural
NNP  → Proper noun, singular
NNPS → Proper noun, plural
```

Example:

```text
dog   → NN
dogs  → NNS
Ali   → NNP
```

---

### Verbs

```text
VB   → Base verb
VBD  → Past tense
VBG  → Gerund / present participle
VBN  → Past participle
VBP  → Present verb
VBZ  → Present verb, 3rd person singular
```

Example:

```text
run      → VB
walked   → VBD
running  → VBG
eaten    → VBN
runs     → VBZ
```

---

### Adjectives

```text
JJ   → Adjective
JJR  → Comparative adjective
JJS  → Superlative adjective
```

Example:

```text
beautiful → JJ
better    → JJR
best      → JJS
```

---

### Adverbs

```text
RB   → Adverb
RBR  → Comparative adverb
RBS  → Superlative adverb
```

Example:

```text
quickly → RB
faster  → RBR
fastest → RBS
```

---

### Pronouns

```text
PRP  → Personal pronoun
PRP$ → Possessive pronoun
```

Example:

```text
he   → PRP
she  → PRP
my   → PRP$
their → PRP$
```

---

### Other Important Tags

```text
DT   → Determiner
IN   → Preposition / subordinating conjunction
CC   → Coordinating conjunction
TO   → to
CD   → Cardinal number
MD   → Modal
UH   → Interjection
```

# 9. POS Tagging Step-by-Step

```text
Sentence
    ↓
Tokenization
    ↓
Individual words
    ↓
POS Tagger
    ↓
Word + POS Tag
```

Example:

```text

      ↓

["The", "boy", "runs"]

      ↓

POS Tagging

      ↓

[
 ("The", "DT"),

]
```

# 12. POS Tagging and Lemmatization

POS tagging can also help with **lemmatization**.

Without POS:

```python
lemmatizer.lemmatize("running")
```

May return:

```text
running
```

Because the default POS is noun.

If we know that `running` is a verb:

```python
lemmatizer.lemmatize("running", pos="v")
```

Output:

```text
run
```

So a more advanced NLP pipeline can be:

```text
Text
 ↓
Tokenization
 ↓
POS Tagging
 ↓
Lemmatization
 ↓
Clean Text
```

---

# 13. Important Point

POS tagging does **not** remove or change the words.

It simply adds information:

```text
running
```

becomes:

```text
(running, VBG)
```

The original word is still there.

---

# 14. Stemming vs Lemmatization vs POS Tagging

### Stemming

```text
studies → studi
```

Reduces a word to a stem.

### Lemmatization

```text
studies → study
```

Reduces a word to its meaningful base form.

### POS Tagging

```text
studies → NNS / VBZ
```

Identifies the grammatical role of the word.
